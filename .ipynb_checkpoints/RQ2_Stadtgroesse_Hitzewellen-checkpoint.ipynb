# RQ2 — Hitzewellen nach Stadtgröße (Deutschland)

**Forschungsfrage:** Wie unterscheiden sich Häufigkeit, Intensität (Peak- und
Durchschnittstemperatur) und Dauer von Hitzewellen zwischen Großstadt, Mittelstadt,
Kleinstadt und Landgemeinde, jeweils in derselben Region?

**Ansatz:** Für jede der deutschen Großstädte (>=150.000 Einwohner) aus RQ1 wird automatisch
die naechstgelegene Mittelstadt, Kleinstadt und Landgemeinde im jeweiligen Einwohnerfenster
gesucht (mind. 8 km entfernt, damit wir nicht in dieselbe Wettermodell-Gitterzelle wie die
Großstadt fallen; max. 50 km, damit wir in derselben Klimaregion bleiben). Anschließend werden
für alle diese Orte dieselben Hitzewellen-Funktionen wie in RQ1 angewendet, und am Ende wird
nach Größenklasse gruppiert ausgewertet.

> **Hinweis zur Ausführung:** Dieses Notebook wurde in einer Sandbox ohne Zugriff auf
> `query.wikidata.org` und `archive-api.open-meteo.com` erstellt. Der Code ist direkt aus der
> in unserer Besprechung abgestimmten Logik aufgebaut und syntaktisch geprüft, konnte hier aber
> nicht live gegen echte Daten getestet werden. Führt die Zellen der Reihe nach in eurer
> eigenen Umgebung aus (Kurztest-Zellen zuerst, wie gewohnt), bevor ihr den vollen 56-Städte-Lauf
> startet. Falls `cities_df` und `CLIENT` aus eurem RQ1-Notebook schon im selben Kernel geladen
> sind, koennt ihr die Zellen 1-3 ueberspringen.

Zelle 1 — Imports

Alles, was wir aus RQ1 und für die neuen Geo-Suchen brauchen.

```
import time
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import openmeteo_requests
import requests_cache
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry
```

Zelle 2 — Großstädte laden (wie in RQ1)

SPARQL-Anfrage bei Wikidata für deutsche Städte >=150.000 Einwohner, inkl. Koordinaten.
Der "Freie Hansestadt Bremen"-Duplikat wird explizit rausgefiltert (dieselbe Stadt wie
"Bremen", nur ein anderer Wikidata-Eintrag für den Stadtstaat).

```
def get_german_cities(min_population=150000):
    query = f"""
    SELECT DISTINCT ?cityLabel ?population ?coord WHERE {{
      ?city wdt:P31/wdt:P279* wd:Q515;      # ist eine Stadt
            wdt:P17 wd:Q183;                # liegt in Deutschland
            wdt:P1082 ?population;          # hat Einwohnerzahl
            wdt:P625 ?coord.                # hat Koordinaten
      FILTER(?population >= {min_population})
      SERVICE wikibase:label {{ bd:serviceParam wikibase:language "de,en". }}
    }}
    ORDER BY DESC(?population)
    """
    url = "https://query.wikidata.org/sparql"
    headers = {"Accept": "application/sparql-results+json", "User-Agent": "student-project/1.0"}
    r = requests.get(url, params={"query": query}, headers=headers)
    r.raise_for_status()

    rows = r.json()["results"]["bindings"]
    cities = []
    for row in rows:
        name = row["cityLabel"]["value"]
        pop = int(row["population"]["value"])
        lon, lat = row["coord"]["value"].replace("Point(", "").replace(")", "").split(" ")
        cities.append({"city": name, "population": pop, "lat": float(lat), "lon": float(lon)})

    df = pd.DataFrame(cities).drop_duplicates(subset="city")
    return df.sort_values("population", ascending=False).reset_index(drop=True)


cities_df = get_german_cities(min_population=150000)

# Bremen-Duplikat entfernen: "Freie Hansestadt Bremen" ist derselbe Ort wie "Bremen"
cities_df = cities_df[cities_df["city"] != "Freie Hansestadt Bremen"].reset_index(drop=True)

print(cities_df)
print(f"Anzahl Grossstaedte: {len(cities_df)}")
```

Zelle 3 — Open-Meteo-Client (wie in RQ1)

Persistenter Cache (läuft nie ab, da sich historische Klimadaten nicht rückwirkend ändern)
plus Retry-Logik inklusive 429 (Rate-Limit).

```
def make_client():
    cache_session = requests_cache.CachedSession(
        "openmeteo_cache",
        expire_after=-1,
        allowable_codes=[200],
    )
    retry = Retry(
        total=6, backoff_factor=3,
        status_forcelist=[429, 500, 502, 503, 504],
        allowed_methods=["GET"],
    )
    cache_session.mount("https://", HTTPAdapter(max_retries=retry))
    cache_session.headers.update({"User-Agent": "student-project/1.0"})
    return openmeteo_requests.Client(session=cache_session)


CLIENT = make_client()
```

Zelle 4 — Wetterdaten laden + Hitzewellen erkennen (wie in RQ1)

`fetch_tmax_batch` holt Tages-Maximaltemperaturen für mehrere Orte gleichzeitig.
`compute_doy_thresholds` berechnet pro Kalendertag einen Schwellenwert (98. Perzentil aus
+-15 Tagen der Referenzperiode). `find_dwd_heatwaves` erkennt Hitzewellen als mindestens 3
Tage in Folge über diesem Klima-Schwellenwert UND über einem festen 28-Grad-Grenzwert.

```
ARCHIVE_URL = "https://archive-api.open-meteo.com/v1/archive"


def fetch_tmax_batch(client, coords, start_date, end_date, batch_size=15, pause=5):
    """Holt Tages-Maximaltemperaturen fuer mehrere Orte gleichzeitig."""
    dfs = []
    n_batches = -(-len(coords) // batch_size)

    for b in range(n_batches):
        i = b * batch_size
        chunk = coords[i:i + batch_size]

        params = {
            "latitude": [lat for lat, lon in chunk],
            "longitude": [lon for lat, lon in chunk],
            "start_date": start_date, "end_date": end_date,
            "daily": ["temperature_2m_max"],
            "timezone": "Europe/Berlin",
        }
        try:
            responses = client.weather_api(ARCHIVE_URL, params=params)
        except Exception as e:
            raise RuntimeError(f"API-Fehler bei Batch {b+1}/{n_batches}: {e}") from e

        for response in responses:
            daily = response.Daily()
            tmax = daily.Variables(0).ValuesAsNumpy()
            dates = pd.date_range(
                start=pd.to_datetime(daily.Time(), unit="s", utc=True),
                end=pd.to_datetime(daily.TimeEnd(), unit="s", utc=True),
                freq=pd.Timedelta(seconds=daily.Interval()),
                inclusive="left",
            ).tz_localize(None)
            dfs.append(pd.DataFrame({"date": dates, "tmax": tmax}))

        print(f"  Batch {b+1}/{n_batches} ok ({len(chunk)} Orte)")
        time.sleep(pause)

    return dfs


def compute_doy_thresholds(ref_df, window=15, percentile=98):
    """Schwellenwert pro Kalendertag: 98. Perzentil aus +-15 Tagen der Referenzperiode."""
    doy = ref_df["doy"].to_numpy()
    tmax = ref_df["tmax"].to_numpy()
    thresholds = {}

    for d in range(1, 366):
        dist = np.abs(doy - d)
        dist = np.minimum(dist, 365 - dist)
        thresholds[d] = np.percentile(tmax[dist <= window], percentile)

    return thresholds


def find_dwd_heatwaves(df, thresholds, min_days=3, fixed_threshold=28.0):
    """Hitzewelle = mind. 3 Tage in Folge ueber Klima-Schwelle UND ueber 28 Grad."""
    df = df.copy()
    df["doy"] = df["date"].dt.dayofyear.clip(upper=365)
    df["threshold"] = df["doy"].map(thresholds)
    df["is_hot"] = (df["tmax"] > df["threshold"]) & (df["tmax"] > fixed_threshold)

    heatwaves, start = [], None
    for i, row in df.iterrows():
        if row["is_hot"] and start is None:
            start = i
        elif not row["is_hot"] and start is not None:
            segment = df.loc[start:i - 1]
            if len(segment) >= min_days:
                heatwaves.append({
                    "start": segment["date"].iloc[0].date(),
                    "end": segment["date"].iloc[-1].date(),
                    "duration_days": len(segment),
                    "max_temp": round(segment["tmax"].max(), 1),
                    "avg_temp": round(segment["tmax"].mean(), 1)
                })
            start = None

    if start is not None:
        segment = df.loc[start:]
        if len(segment) >= min_days:
            heatwaves.append({
                "start": segment["date"].iloc[0].date(),
                "end": segment["date"].iloc[-1].date(),
                "duration_days": len(segment),
                "max_temp": round(segment["tmax"].max(), 1),
                "avg_temp": round(segment["tmax"].mean(), 1)
            })

    return heatwaves


def analyze_city_heatwaves_from_data(ref_df, analysis_df):
    """Verbindet Schwellenwert-Berechnung und Hitzewellen-Suche fuer einen Ort."""
    ref_df = ref_df.copy()
    ref_df["doy"] = ref_df["date"].dt.dayofyear.clip(upper=365)
    thresholds = compute_doy_thresholds(ref_df)
    return find_dwd_heatwaves(analysis_df, thresholds)
```

Zelle 5 — Nachbarort-Suche über Wikidata (neu für RQ2)

Findet den naechstgelegenen Ort in einem Populations-Fenster. `min_dist_km` verhindert, dass
wir in dieselbe Open-Meteo-Gitterzelle wie die Referenz-Großstadt fallen (Auflösung dort ca.
5-25 km je nach Zeitraum); `max_dist_km` haelt uns in derselben Klimaregion.

```
WD_URL = "https://query.wikidata.org/sparql"
WD_HEADERS = {"Accept": "application/sparql-results+json", "User-Agent": "student-project/1.0"}


def find_nearby_settlement(lat, lon, min_pop, max_pop, min_dist_km=8, max_dist_km=50, radius_km=50):
    """Naechstgelegene deutsche Siedlung in einem Populations-Fenster."""
    query = f"""
    PREFIX wd: <http://www.wikidata.org/entity/>
    PREFIX wdt: <http://www.wikidata.org/prop/direct/>
    PREFIX bd: <http://www.bigdata.com/rdf#>
    PREFIX wikibase: <http://wikiba.se/ontology#>
    PREFIX geo: <http://www.opengis.net/ont/geosparql#>
    PREFIX geof: <http://www.opengis.net/def/function/geosparql/>

    SELECT ?placeLabel ?population ?coord ?dist WHERE {{
      SERVICE wikibase:around {{
        ?place wdt:P625 ?coord .
        bd:serviceParam wikibase:center "Point({lon} {lat})"^^geo:wktLiteral .
        bd:serviceParam wikibase:radius "{radius_km}" .
      }}
      BIND(geof:distance(?coord, "Point({lon} {lat})"^^geo:wktLiteral) AS ?dist)
      ?place wdt:P31/wdt:P279* wd:Q486972 .   # menschliche Siedlung (Oberbegriff)
      ?place wdt:P17 wd:Q183 .                 # liegt in Deutschland
      ?place wdt:P1082 ?population .
      FILTER(?population >= {min_pop} && ?population < {max_pop})
      FILTER(?dist >= {min_dist_km} && ?dist <= {max_dist_km})
      SERVICE wikibase:label {{ bd:serviceParam wikibase:language "de,en". }}
    }}
    ORDER BY ?dist
    LIMIT 5
    """
    r = requests.get(WD_URL, params={"query": query}, headers=WD_HEADERS, timeout=30)
    r.raise_for_status()
    rows = r.json()["results"]["bindings"]
    if not rows:
        return None
    row = rows[0]
    name = row["placeLabel"]["value"]
    pop = int(row["population"]["value"])
    dist = float(row["dist"]["value"])
    lon_r, lat_r = row["coord"]["value"].replace("Point(", "").replace(")", "").split(" ")
    return {"place": name, "population": pop, "lat": float(lat_r), "lon": float(lon_r), "distance_km": round(dist, 1)}


SIZE_BINS = {
    "Mittelstadt": (20_000, 150_000),   # direkt unter der Grossstadt-Schwelle
    "Kleinstadt": (5_000, 20_000),
    "Landgemeinde": (200, 5_000),       # 200 als Untergrenze gegen Mini-Weiler mit kaum Daten
}
```

Zelle 6 — Kurztest: 3 Großstädte

Prüft, ob `find_nearby_settlement` sinnvolle Orte zurückgibt, bevor der volle Lauf über alle
Großstädte startet — genau wie der Kurztest-Ansatz aus RQ1.

```
for _, row in cities_df.head(3).iterrows():
    print(f"\n{row['city']}:")
    for size_class, (min_pop, max_pop) in SIZE_BINS.items():
        match = find_nearby_settlement(row["lat"], row["lon"], min_pop, max_pop)
        print(f"  {size_class}: {match}")
    time.sleep(1)
```

Zelle 7 — Volle Schleife über alle Großstädte

Baut `locations_df`: pro Großstadt eine Zeile "Grossstadt" plus je eine Zeile für die
gefundene Mittelstadt/Kleinstadt/Landgemeinde. Orte, für die nichts Passendes gefunden wurde,
landen in `skipped` (analog zu `failed_cities` aus RQ1).

```
reference_rows = []
skipped = []

for idx, row in cities_df.iterrows():
    reference_rows.append({
        "cluster": row["city"], "size_class": "Grossstadt",
        "place": row["city"], "population": row["population"],
        "lat": row["lat"], "lon": row["lon"], "distance_km": 0.0,
    })
    for size_class, (min_pop, max_pop) in SIZE_BINS.items():
        try:
            match = find_nearby_settlement(row["lat"], row["lon"], min_pop, max_pop)
        except Exception as e:
            print(f"  Fehler bei {row['city']} / {size_class}: {e}")
            match = None
        if match is None:
            skipped.append((row["city"], size_class))
        else:
            reference_rows.append({"cluster": row["city"], "size_class": size_class, **match})
    time.sleep(1)

locations_df = pd.DataFrame(reference_rows)
print(locations_df)
print(f"\nNicht gefunden: {len(skipped)} von {len(cities_df) * 3}")
print(skipped)
```

Zelle 8 — Wetterdaten für alle Orte in `locations_df`

Referenzperiode 1961-1990 (WMO-Klimareferenz) und ein Analysezeitraum. Die 56 Großstädte
kommen dabei sofort aus dem Cache (schon in RQ1 abgerufen), nur für die neuen Orte werden
wirklich neue Daten geholt.

`ANALYSIS_START`/`ANALYSIS_END` unten ist aktuell auf 2025 gesetzt (passend zu eurem
RQ1-Stand). Für robustere Mittelwerte pro Größenklasse könnt ihr das auf mehrere Jahre
ausweiten (z. B. 2016-01-01 bis 2025-12-31) — dauert dann proportional länger.

```
ANALYSIS_START = "2025-01-01"
ANALYSIS_END = "2025-12-31"

coords_all = list(zip(locations_df["lat"], locations_df["lon"]))

ref_dfs_all = fetch_tmax_batch(CLIENT, coords_all, "1961-01-01", "1990-12-31", batch_size=15, pause=5)

time.sleep(120)  # Puffer, damit sich das Rate-Limit-Fenster erholt

analysis_dfs_all = fetch_tmax_batch(CLIENT, coords_all, ANALYSIS_START, ANALYSIS_END, batch_size=15, pause=5)
```

Zelle 9 — Hitzewellen pro Ort berechnen

```
results = []
failed = []

for idx, row in locations_df.iterrows():
    try:
        heatwaves = analyze_city_heatwaves_from_data(ref_dfs_all[idx], analysis_dfs_all[idx])
        for hw in heatwaves:
            results.append({
                "cluster": row["cluster"], "size_class": row["size_class"],
                "place": row["place"], "population": row["population"],
                **hw
            })
    except Exception as e:
        print(f"  Uebersprungen wegen Fehler ({row['place']}): {e}")
        failed.append(row["place"])

results_df2 = pd.DataFrame(results)
print(results_df2)
print(f"\nFehlgeschlagen: {failed}")
```

Zelle 10 — Auswertung nach Größenklasse

Häufigkeit (Anzahl Hitzewellen pro Ort im Schnitt), Ø-Dauer, Ø-Maximaltemperatur und
Ø-Durchschnittstemperatur je Größenklasse.

```
n_places_per_class = locations_df.groupby("size_class")["place"].nunique()

summary = results_df2.groupby("size_class").agg(
    anzahl_hitzewellen=("place", "count"),
    mean_duration=("duration_days", "mean"),
    mean_max_temp=("max_temp", "mean"),
    mean_avg_temp=("avg_temp", "mean"),
).reindex(["Grossstadt", "Mittelstadt", "Kleinstadt", "Landgemeinde"])

summary["hitzewellen_pro_ort"] = summary["anzahl_hitzewellen"] / n_places_per_class
summary = summary.round(2)
print(summary)
```

Zelle 11 — Visualisierung

```
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

summary["hitzewellen_pro_ort"].plot(kind="bar", ax=axes[0], color="steelblue")
axes[0].set_title("Ø Hitzewellen pro Ort")
axes[0].set_xlabel("")

summary["mean_duration"].plot(kind="bar", ax=axes[1], color="darkorange")
axes[1].set_title("Ø Dauer (Tage)")
axes[1].set_xlabel("")

summary["mean_max_temp"].plot(kind="bar", ax=axes[2], color="firebrick")
axes[2].set_title("Ø Maximaltemperatur (°C)")
axes[2].set_xlabel("")

plt.tight_layout()
plt.show()
```
