# Are Heatwaves in Germany Really Becoming More Common? A Data Investigation

**The question:** Have German major cities experienced more, longer, and hotter heatwaves since 1980?

**The approach:** We look at 57 German cities with more than 150,000 inhabitants and analyze 45 years of daily maximum temperatures for each one (1980–2025). For every single day, we check two things: was it unusually hot **for that specific city** (compared to the years 1961–1990), and was it hot in absolute terms (above 28°C)? Only if at least 3 days in a row meet both conditions does it count as a heatwave.

This notebook walks through the method step by step — each step comes with a chart, so it stays clear what's actually happening, before everything is scaled up to all 57 cities at the end.

Cell 2 — Which cities are we looking at?

57 German cities with more than 150,000 inhabitants, loaded via a Wikidata query (a single, lightweight request — no cache needed). Instead of just a table: a simple map where dot size shows population. Berlin, Hamburg, and Munich stand out at a glance as the three largest dots.

```python
# Cell 3
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def get_german_cities(min_population=150000):
    query = f"""
    SELECT DISTINCT ?cityLabel ?population ?coord WHERE {{
      ?city wdt:P31/wdt:P279* wd:Q515;
            wdt:P17 wd:Q183;
            wdt:P1082 ?population;
            wdt:P625 ?coord.
      FILTER(?population >= {min_population})
      SERVICE wikibase:label {{ bd:serviceParam wikibase:language "de,en". }}
    }}
    ORDER BY DESC(?population)
    """
    url = "https://query.wikidata.org/sparql"
    headers = {"Accept": "application/sparql-results+json", "User-Agent": "student-project/1.0"}
    r = requests.get(url, params={"query": query}, headers=headers)
    r.raise_for_status()

    rows = r.json()["results"]["bindings"]
    cities = []
    for row in rows:
        name = row["cityLabel"]["value"]
        pop = int(row["population"]["value"])
        lon, lat = row["coord"]["value"].replace("Point(", "").replace(")", "").split(" ")
        cities.append({"city": name, "population": pop, "lat": float(lat), "lon": float(lon)})

    df = pd.DataFrame(cities).drop_duplicates(subset="city")
    return df.sort_values("population", ascending=False).reset_index(drop=True)

cities_df = get_german_cities(min_population=150000)

# Translate only the handful of city names that are genuinely different in English.
# (Wikidata's English label service was tried first, but it makes "Freie Hansestadt
# Bremen" and "Bremen" collide into the same label, silently dropping the real city --
# so this list is applied manually on top of the reliable German-language query.)
english_names = {
    "München": "Munich", "Köln": "Cologne", "Nürnberg": "Nuremberg",
    "Freie Hansestadt Bremen": "Bremen (state)",
}
cities_df["city"] = cities_df["city"].replace(english_names)

print(f"{len(cities_df)} cities loaded")

total_all = cities_df["population"].sum()
total_excl = cities_df.drop(index=5)["population"].sum()  # excluding "Bremen (state)" -- see exclude_idx
print(f"Total population across all {len(cities_df)} cities: {total_all:,}")
print(f"Total population excluding the Bremen duplicate ({len(cities_df) - 1} cities): {total_excl:,}")

# Germany outline (public GeoJSON, low resolution -- plenty for an overview map)
geo_url = "https://raw.githubusercontent.com/isellsoap/deutschlandGeoJSON/main/1_deutschland/4_niedrig.geo.json"
germany_outline = requests.get(geo_url, timeout=30).json()

fig, ax = plt.subplots(figsize=(7, 9))

for polygon in germany_outline["features"][0]["geometry"]["coordinates"]:
    for ring in polygon:
        ring = np.array(ring)
        ax.plot(ring[:, 0], ring[:, 1], color="black", linewidth=0.8)
        ax.fill(ring[:, 0], ring[:, 1], color="whitesmoke", zorder=0)

ax.scatter(cities_df["lon"], cities_df["lat"], s=cities_df["population"] / 3000,
           alpha=0.7, color="darkred", edgecolor="black", linewidth=0.5, zorder=3)
for _, row in cities_df.head(8).iterrows():
    ax.annotate(row["city"], (row["lon"], row["lat"]), fontsize=10,
                xytext=(5, 5), textcoords="offset points", zorder=4)

ax.set_title("The 57 German major cities studied\n(dot size = population)",
              fontsize=14, fontweight="bold")
ax.set_aspect("equal")
ax.axis("off")
plt.savefig("data/processed/rq1_map_cities.svg")
plt.savefig("data/processed/rq1_map_cities.png", dpi=400)
plt.tight_layout()
plt.show()
```

Cell 4 — Loading everything from local files

Two things were already computed and saved in the main project: the detected heatwaves (`data/processed/heatwaves.csv`) and the raw daily temperatures for 1980–2025 (`data/raw/tmax_1980_2025.parquet`), plus the 1961–1990 reference period (`data/raw/tmax_1961_1990.parquet`). This notebook only reads these three files — it never touches the API.

```python
# Cell 5
import pandas as pd

exclude_idx = {5}  # "Free Hanseatic City of Bremen" -- a federal-state entry, not the city itself

results_df = pd.read_csv("data/processed/heatwaves.csv", parse_dates=["start", "end"])
tmax_df = pd.read_parquet("data/raw/tmax_1980_2025.parquet")
ref_df = pd.read_parquet("data/raw/tmax_1961_1990.parquet")

print(f"{len(results_df)} heatwaves loaded, covering {results_df['city'].nunique()} cities")
```

Cell 6 — One function we still need

Everything else (fetching, detecting) is already done and loaded above. The only piece of the original methodology worth keeping here is the calculation behind the threshold curve in Cell 11 — it's the clearest way to show *why* a day counts as "unusually hot" for a specific city.

```python
# Cell 7
import numpy as np

def compute_doy_thresholds(ref_city_df, window=15, percentile=98):
    doy = ref_city_df["date"].dt.dayofyear.clip(upper=365).to_numpy()
    tmax = ref_city_df["tmax"].to_numpy()
    thresholds = {}
    for d in range(1, 366):
        dist = np.minimum(np.abs(doy - d), 365 - np.abs(doy - d))
        thresholds[d] = np.percentile(tmax[dist <= window], percentile)
    return thresholds
```

Cell 8 — Example: what does the raw data actually look like?

The daily maximum temperature for **Berlin in 2003** — chosen deliberately, because that year includes the famous European heatwave. The dashed line at 28°C is the fixed threshold from the definition.

```python
# Cell 9
import matplotlib.pyplot as plt

berlin_2003 = tmax_df[(tmax_df["city"] == "Berlin") & (tmax_df["date"].dt.year == 2003)]

fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(berlin_2003["date"], berlin_2003["tmax"], color="steelblue", linewidth=1.2)
ax.axhline(28, color="darkred", linestyle="--", label="28°C threshold")
ax.set_title("Berlin, 2003: daily maximum temperature", fontsize=14, fontweight="bold")
ax.set_ylabel("Temperature (°C)", fontsize=11)
ax.legend()
ax.grid(alpha=0.3)
plt.savefig("data/processed/rq1_berlin2003_daily.svg")
plt.savefig("data/processed/rq1_berlin2003_daily.png", dpi=400)
plt.tight_layout()
plt.show()
```

Cell 10 — What counts as "unusually hot" for this city?

The orange line shows, for every day of the year, how hot that day typically was for Berlin during 1961–1990 (98th percentile). The blue line is 2003 again — only where it lies above the orange line does a day count as unusually hot for Berlin specifically.

```python
# Cell 11
ref_berlin = ref_df[ref_df["city"] == "Berlin"]
thresholds_berlin = compute_doy_thresholds(ref_berlin)
threshold_curve = pd.Series(thresholds_berlin).sort_index()

berlin_2003_doy = berlin_2003.copy()
berlin_2003_doy["doy"] = berlin_2003_doy["date"].dt.dayofyear

fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(threshold_curve.index, threshold_curve.values, color="orange", linewidth=2,
        label="Threshold for Berlin (from 1961–1990)")
ax.plot(berlin_2003_doy["doy"], berlin_2003_doy["tmax"], color="steelblue", linewidth=1, alpha=0.8,
        label="Actual temperature in 2003")
ax.set_title("When was a day in Berlin in 2003 'unusually hot'?", fontsize=14, fontweight="bold")
ax.set_xlabel("Day of year (1 = January 1st, 365 = December 31st)", fontsize=11)
ax.set_ylabel("Temperature (°C)", fontsize=11)
ax.legend()
ax.grid(alpha=0.3)
plt.savefig("data/processed/rq1_berlin2003_threshold.svg")
plt.savefig("data/processed/rq1_berlin2003_threshold.png", dpi=400)
plt.tight_layout()
plt.show()
```

Cell 12 — A real heatwave, up close

This event was already detected and saved in `heatwaves.csv` — no need to recompute it. We just pick it out and zoom into the surrounding days from the raw temperature data.

```python
# Cell 13
berlin_2003_events = results_df[(results_df["city"] == "Berlin") & (results_df["start"].dt.year == 2003)]
longest = berlin_2003_events.sort_values("duration_days", ascending=False).iloc[0]

window_start = longest["start"] - pd.Timedelta(days=7)
window_end = longest["end"] + pd.Timedelta(days=7)
berlin_all = tmax_df[tmax_df["city"] == "Berlin"]
window = berlin_all[(berlin_all["date"] >= window_start) & (berlin_all["date"] <= window_end)]

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(window["date"], window["tmax"], marker="o", color="steelblue")
ax.axhline(28, color="gray", linestyle="--", alpha=0.7, label="28°C threshold")
ax.axvspan(longest["start"], longest["end"], color="red", alpha=0.2, label="Detected heatwave")
ax.set_title(f"Berlin: heatwave from {longest['start'].date()} to {longest['end'].date()} "
             f"({longest['duration_days']} days, peak {longest['max_temp']}°C)",
             fontsize=14, fontweight="bold")
ax.set_ylabel("Temperature (°C)", fontsize=11)
ax.legend()
ax.grid(alpha=0.3)
plt.savefig("data/processed/rq1_berlin_heatwave_zoom.svg")
plt.savefig("data/processed/rq1_berlin_heatwave_zoom.png", dpi=400)
plt.tight_layout()
plt.show()
```

Cell 14 — Now scale that up

`results_df`, loaded in Cell 5, already contains this exact same process — repeated automatically for every one of the 57 cities across all 46 years.

Cell 18 — The big picture: how has this developed over 45 years?

Four metrics per year, averaged or summed across all 57 cities: how often, how hot, how long.

```python
# Cell 19
all_years = pd.DataFrame({"year": range(1980, 2026)})
yearly = all_years.merge(
    results_df.groupby("year").agg(
        n_heatwaves=("city", "count"), avg_max_temp=("max_temp", "mean"),
        avg_avg_temp=("avg_temp", "mean"), avg_duration=("duration_days", "mean"),
    ).reset_index(), on="year", how="left")
yearly["n_heatwaves"] = yearly["n_heatwaves"].fillna(0).astype(int)

fig, axes = plt.subplots(2, 2, figsize=(16, 9))
plots = [("n_heatwaves", "Number of heatwaves (all 57 cities)", axes[0,0], "steelblue"),
         ("avg_max_temp", "Avg. peak temperature (°C)", axes[0,1], "darkred"),
         ("avg_avg_temp", "Avg. mean temperature (°C)", axes[1,0], "orange"),
         ("avg_duration", "Avg. duration (days)", axes[1,1], "purple")]
for col, label, ax, color in plots:
    ax.plot(yearly["year"], yearly[col], marker="o", color=color)
    ax.set_title(label, fontsize=13, fontweight="bold")
    ax.grid(alpha=0.3)
plt.savefig("data/processed/rq1_yearly_overview.svg")
plt.savefig("data/processed/rq1_yearly_overview.png", dpi=400)
plt.tight_layout()
plt.show()
```

Cell 20 — Is this just random, or a real trend?

The ups and downs in the charts above could just be random noise. A linear regression checks whether there's a genuine direction over the 45 years. Green bars = statistically confirmed trend (the probability that this is random chance is under 5%), gray bars = no reliable evidence.

```python
# Cell 21
from scipy import stats

metrics = [("n_heatwaves", "Frequency\n(heatwaves/year)"), ("avg_max_temp", "Peak temperature\n(°C/year)"),
           ("avg_avg_temp", "Mean temperature\n(°C/year)"), ("avg_duration", "Duration\n(days/year)")]

slopes, colors, labels = [], [], []
for col, label in metrics:
    valid = yearly.dropna(subset=[col])
    slope, _, _, p, _ = stats.linregress(valid["year"], valid[col])
    slopes.append(slope)
    colors.append("seagreen" if p < 0.05 else "lightgray")
    labels.append(f"{label}\n(p={p:.3f})")

fig, ax = plt.subplots(figsize=(9, 6))
ax.bar(labels, slopes, color=colors)
ax.axhline(0, color="black", linewidth=0.8)
ax.set_title("Change per year — green = statistically confirmed trend", fontsize=14, fontweight="bold")
ax.grid(axis="y", alpha=0.3)
plt.savefig("data/processed/rq1_significance.svg")
plt.savefig("data/processed/rq1_significance.png", dpi=400)
plt.tight_layout()
plt.show()
```

## Conclusion

**Yes — heatwaves in Germany have measurably become more frequent and longer since 1980.** Across all 57 German major cities, the number of heatwaves per year has risen with statistical confidence, as has their average duration. The picture for peak temperature is more mixed: over the full 1980–2025 period (partly due to fewer comparable years available), no reliable trend can be shown — restricting the comparison to the years from 1991 onward (to exclude any overlap with the 1961–1990 baseline period) makes this trend statistically visible too.

**In short:** heatwaves are becoming more frequent and lasting longer — but the individual heatwave itself isn't becoming dramatically hotter than before.

```python
# Poster chart: replaces the two line plots with decade-binned bars
def decade_bin(year):
    if year < 1990: return "1980–89"
    elif year < 2000: return "1990–99"
    elif year < 2010: return "2000–09"
    elif year < 2020: return "2010–19"
    else: return "2020–25"

yearly["decade"] = yearly["year"].apply(decade_bin)
decade_order = ["1980–89", "1990–99", "2000–09", "2010–19", "2020–25"]

decade_stats = yearly.groupby("decade").agg(
    freq_mean=("n_heatwaves", "mean"), freq_std=("n_heatwaves", "std"),
    dur_mean=("avg_duration", "mean"), dur_std=("avg_duration", "std"),
).reindex(decade_order)

fig, axes = plt.subplots(1, 2, figsize=(10, 4.5))

axes[0].bar(decade_order, decade_stats["freq_mean"], yerr=decade_stats["freq_std"],
            color="steelblue", capsize=4)
axes[0].set_title("Number of heatwaves per year", fontsize=13, fontweight="bold")
axes[0].set_ylabel("Heatwaves/year (avg.)", fontsize=11)
axes[0].tick_params(axis="x", rotation=30)
axes[0].grid(axis="y", alpha=0.3)

axes[1].bar(decade_order, decade_stats["dur_mean"], yerr=decade_stats["dur_std"],
            color="purple", capsize=4)
axes[1].set_title("Avg. duration per heatwave", fontsize=13, fontweight="bold")
axes[1].set_ylabel("Duration (days, avg.)", fontsize=11)
axes[1].tick_params(axis="x", rotation=30)
axes[1].grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("data/processed/poster_decade_bars.png", dpi=400)
plt.savefig("data/processed/poster_decade_bars.svg")  # Vektorgrafik -- für PowerPoint, bleibt bei jeder Größe scharf
plt.show()
```
