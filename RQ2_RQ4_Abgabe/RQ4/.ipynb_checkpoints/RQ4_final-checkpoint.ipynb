## RQ2 (Italien)

How do heatwave frequency, intensity (peak and average temperature), and duration differ across city sizes, from large cities to rural municipalities, within the same region in Italy?

Diese Version wendet exakt dieselbe Methodik wie die Deutschland-Analyse (`RQ2.ipynb`) auf Italien an — gleiche Hitzewellen-Definition, gleiche Bevölkerungsschwellen, gleicher Ablauf. Angepasst werden nur die länderspezifischen Teile: die Wikidata-Abfrage (Land, italienische Gemeinden/"comuni" statt deutscher Gemeinden, ISTAT-Gemeindeschlüssel statt amtlichem deutschen Gemeindeschlüssel) und die Kartengrundlage (italienische Regionen statt deutscher Bundesländer).

**Wichtige methodische Anpassung:** Deutschland unterscheidet Mittelstadt/Kleinstadt von Gemeinde zusätzlich über die Wikidata-Markierung "Stadt" (angelehnt an das deutsche Stadtrecht). Italien kennt dieses Rechtskonzept nicht — alle italienischen Orte sind rechtlich gleichgestellte "comuni", unabhängig von Größe oder Charakter. Die Größenklassen werden hier deshalb **ausschließlich über die Einwohnerzahl** bestimmt, ohne zusätzliche Stadt/Land-Filterung. Das ist die methodisch sauberere Lösung für Italien, nicht nur ein technischer Kompromiss.

**Teil 1 (Abschnitte 1-7):** Für **jede Großstadt** wird gesucht nach:

1. der nächstgelegenen **Mittelstadt** mit 20.000–149.999 Einwohnern,
2. der nächstgelegenen **Kleinstadt** mit 5.000–19.999 Einwohnern,
3. der nächstgelegenen **Gemeinde** mit 1–4.999 Einwohnern.

**Teil 2 (Abschnitte 8+):** Auf alle gefundenen Vergleichsorte wird dieselbe Hitzewellen-Definition angewendet wie in RQ2.ipynb (Deutschland) — bewusst unverändert, damit beide Länder methodisch vergleichbar bleiben — und die Ergebnisse werden nach Größenklasse (Großstadt/Mittelstadt/Kleinstadt/Gemeinde) verglichen.

```python
import requests
import json
import time
import pandas as pd
import numpy as np

pd.set_option("display.max_rows", 200)
pd.set_option("display.max_columns", 20)
```

## 1. Alle italienischen Gemeinden (comuni) auf einmal laden
### Note
Wie in der Deutschland-Version: alle italienischen Gemeinden ("comuni") einmal komplett laden und lokal filtern, statt für jede Suche eine neue Datenbankanfrage zu schicken. Die Anfrage ist auch hier zu groß für einen einzelnen Abruf, daher wieder adaptives Laden mit automatischem Splitten bei Fehlern.

**Anpassungen gegenüber der Deutschland-Version:**

- Land: `wd:Q38` (Italien) statt `wd:Q183` (Deutschland).
- Gemeinde-Klasse: `wd:Q747074` ("comune of Italy") statt `wd:Q262166` ("eigenständige deutsche Gemeinde").
- Amtlicher Schlüssel: `wdt:P635` (ISTAT-Gemeindecode, das italienische Äquivalent zum deutschen Gemeindeschlüssel) statt `wdt:P439`.
- **Kein `isCity`-Flag mehr:** Das deutsche Stadtrecht-Konzept (`wd:Q515`) existiert in Italien nicht — alle Gemeinden sind rechtlich gleichgestellte "comuni". Die Größenklassen werden für Italien deshalb ausschließlich über die Einwohnerzahl bestimmt (siehe Abschnitte 3-5).

**Wie der Code das umsetzt:**

- `_fetch_population_range(min_pop, max_pop)` schickt genau eine SPARQL-Anfrage an Wikidata und holt alle italienischen Gemeinden, deren Einwohnerzahl im angegebenen Bereich liegt (Name, Einwohnerzahl, Koordinaten, ISTAT-Gemeindecode). `min_pop`/`max_pop` werden direkt in die Anfrage eingesetzt.
- `_fetch_population_range_adaptive(...)` ruft `_fetch_population_range` auf und fängt HTTP-/Timeout-/JSON-Fehler ab. Schlägt eine Anfrage fehl (meist weil der Bereich zu viele Gemeinden umfasst und die Antwort für den Wikidata-Server zu groß wird), wird der Populationsbereich in der Mitte halbiert und beide Hälften rekursiv einzeln angefragt. `max_depth` verhindert, dass sich das unendlich oft wiederholt.
- `get_all_italian_municipalities(...)` ruft die adaptive Funktion einmal für den gesamten Populationsbereich auf, wandelt das Ergebnis in ein DataFrame um und entfernt Duplikate über `municipality_key` (den ISTAT-Gemeindecode, eindeutiger als der Ortsname).

Das Ergebnis (`all_places_df`) wird genau einmal geladen und dann im gesamten weiteren Notebook lokal gefiltert und durchsucht (Abschnitte 2-5), statt für jede Großstadt einzeln eine neue, langsame Datenbankanfrage an Wikidata zu schicken.

```python
WIKIDATA_URL = "https://query.wikidata.org/sparql"

HEADERS = {
    "Accept": "application/sparql-results+json",
    "User-Agent": "student-climate-project/1.0"
}


def _fetch_population_range(min_pop, max_pop):
    query = f"""
    SELECT DISTINCT ?place ?placeLabel ?population ?coord ?municipalityKey WHERE {{
      ?place wdt:P17 wd:Q38;                       # liegt in Italien
             wdt:P31/wdt:P279* wd:Q747074;          # comune of Italy
             wdt:P1082 ?population;                 # hat Einwohnerzahl
             wdt:P625 ?coord;                       # hat Koordinaten
             wdt:P635 ?municipalityKey.              # ISTAT-Gemeindecode

      FILTER(?population >= {min_pop} && ?population <= {max_pop})

      SERVICE wikibase:label {{ bd:serviceParam wikibase:language "it,en". }}
    }}
    """

    r = requests.get(WIKIDATA_URL, params={"query": query}, headers=HEADERS, timeout=180)
    r.raise_for_status()

    payload = json.loads(r.text, strict=False)
    return payload["results"]["bindings"]


def _fetch_population_range_adaptive(min_pop, max_pop, depth=0, max_depth=10):
    try:
        rows = _fetch_population_range(min_pop, max_pop)
        print(f"  ... Population {min_pop}-{max_pop}: {len(rows)} Gemeinden")
        return rows
    except (requests.HTTPError, requests.exceptions.Timeout, json.JSONDecodeError) as e:
        if depth >= max_depth or max_pop <= min_pop:
            raise RuntimeError(
                f"Populationsbereich {min_pop}-{max_pop} lässt sich auch nach {depth} "
                f"Splits nicht laden. Original-Fehler: {e}"
            ) from e

        mid = (min_pop + max_pop) // 2
        print(f"  ... Population {min_pop}-{max_pop} fehlgeschlagen ({e}), splitte bei {mid}")
        time.sleep(1)

        left = _fetch_population_range_adaptive(min_pop, mid, depth + 1, max_depth)
        right = _fetch_population_range_adaptive(mid + 1, max_pop, depth + 1, max_depth)
        return left + right


def get_all_italian_municipalities(min_population=1, max_population=4_000_000):

    rows = _fetch_population_range_adaptive(min_population, max_population)

    places = []
    for row in rows:
        name = row["placeLabel"]["value"]
        pop = int(float(row["population"]["value"]))
        lon, lat = row["coord"]["value"].replace("Point(", "").replace(")", "").split(" ")
        places.append({
            "place": name,
            "wikidata_id": row["place"]["value"].split("/")[-1],
            "population": pop,
            "lat": float(lat),
            "lon": float(lon),
            "municipality_key": row["municipalityKey"]["value"],
        })

    # ISTAT-Gemeindecode ist eindeutiger als nur der Name
    df = pd.DataFrame(places).drop_duplicates(subset="municipality_key")
    return df.reset_index(drop=True)


all_places_df = get_all_italian_municipalities(min_population=1)

print(f"Insgesamt geladen: {len(all_places_df)} italienische Gemeinden (comuni)")
all_places_df.head()
```

## 2. Großstädte aus dem Datensatz herausfiltern
### Note
Hier filtern wir alle italienischen Gemeinden mit Einwohnerzahl >= 150000 — dieselbe Schwelle wie in der Deutschland-Version (siehe die "Wichtige methodische Anpassung" oben: die Schwelle bleibt bewusst identisch, nur die zusätzliche Stadtrecht-Filterung entfällt für Italien). `cities_df` bildet damit die Ausgangsliste für den gesamten weiteren Ablauf: Für jede dieser Großstädte wird in den Abschnitten 3-5 anschließend eine passende Mittelstadt, Kleinstadt und Gemeinde in der Nähe gesucht. Zur besseren Lesbarkeit der Ausgabe wird zusätzlich nach Einwohnerzahl absteigend sortiert.

```python
cities_df = (
    all_places_df[all_places_df["population"] >= 150000]
    .sort_values("population", ascending=False)
    .reset_index(drop=True)
)

print(cities_df)
print(f"Anzahl Großstädte: {len(cities_df)}")
```

## 3. Funktion: Orte in der Nähe einer Großstadt suchen (jetzt lokal)

`find_places_near_city` nutzt die `haversine_km` Funktion, die den Abstand in Km zwischen zwei punkte berechnet und gibt alle Orte, die in der Nähe einer Großstadt sind.

### Note

- `haversine_km(lat1, lon1, lat2, lon2)` berechnet die Großkreis-Entfernung zwischen zwei Punkten auf der Erdkugel in Kilometern (Erdradius `R = 6371.0088`). Alle vier Argumente können auch ganze Numpy-Arrays sein — dann wird die Entfernung eines Punktes zu vielen Punkten gleichzeitig berechnet, ohne eine Python-Schleife (deutlich schneller bei tausenden Orten).
- `find_places_near_city(lat, lon, min_pop, max_pop, radius_km)` filtert `all_places_df` nach Einwohnerzahl `[min_pop, max_pop]` (anders als in der Deutschland-Version **ohne** vorherige Filterung nach Ortstyp — Italien kennt keine Stadtrecht-Unterscheidung, die Größenklasse ergibt sich hier ausschließlich aus der Einwohnerzahl, siehe die "Wichtige methodische Anpassung" ganz oben). Für alle verbleibenden Orte wird die Entfernung zu `(lat, lon)` berechnet und alles außerhalb von `radius_km` verworfen. Das Ergebnis ist nach Entfernung sortiert (nächstgelegener Ort zuerst), damit Abschnitt 4 einfach nur die erste Zeile nehmen muss.

```python
def haversine_km(lat1, lon1, lat2, lon2):
    R = 6371.0088
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    return 2 * R * np.arcsin(np.sqrt(a))


def find_places_near_city(
    lat,
    lon,
    min_pop,
    max_pop=None,
    radius_km=50
):

    df = all_places_df[all_places_df["population"] >= min_pop]
    if max_pop is not None:
        df = df[df["population"] <= max_pop]

    if df.empty:
        return df

    distances = haversine_km(lat, lon, df["lat"].to_numpy(), df["lon"].to_numpy())

    result = df.copy()
    result["distance_km"] = distances
    result = result[result["distance_km"] <= radius_km]

    if result.empty:
        return result

    return (
        result.drop_duplicates(subset="municipality_key")
              .sort_values("distance_km")
              .reset_index(drop=True)
    )
```

## 4. Ermittlung des nächstgelegenen Vergleichsortes
Die Funktion sucht den nächstgelegenen passenden Vergleichsort. Falls kein geeigneter Ort gefunden wird, wird der Suchradius schrittweise erweitert und wenn keinen Ort nach den Radius `radii = [30, 50, 75, 100, 150, 200]` gefunden wurde, dann gibt None zurück.

### Note

`get_nearest_comparison_place` ruft für jeden Radius aus `radii` `find_places_near_city` (Abschnitt 3) auf. Aus den gefundenen Orten werden zwei Fälle herausgefiltert, bevor der erste (= nächstgelegene) übrig bleibt genommen wird: die Großstadt selbst (falls sie zufällig auch die Populations-Kriterien erfüllt) und — bei Gemeinden — Orte, die näher als `min_distance_km` liegen (siehe `RURAL_MIN_DISTANCE_KM` in Abschnitt 5: verhindert, dass ein direkt an die Großstadt angrenzender Vorort fälschlich als "ländliche Gemeinde" durchgeht). Der gefundene `search_radius_km` wird mit zurückgegeben, damit später nachvollziehbar bleibt, wie weit gesucht werden musste.

```python
def get_nearest_comparison_place(
    lat,
    lon,
    min_pop,
    max_pop,
    large_city_name,
    min_distance_km=0
):
    radii = [30, 50, 75, 100, 150, 200]

    for radius in radii:
        df = find_places_near_city(
            lat=lat,
            lon=lon,
            min_pop=min_pop,
            max_pop=max_pop,
            radius_km=radius
        )

        if not df.empty:
            df = df[
                (df["place"].str.lower() != large_city_name.lower()) &
                (df["distance_km"] >= min_distance_km)
            ].copy()

            if not df.empty:
                result = df.iloc[0].copy()
                result["search_radius_km"] = radius
                return result

    return None
```

## 5. Für jede Großstadt Medium, Small und Rural bestimmen

Für jede Großstadt wird eine Mittelstadt, Kleinstadt und Gemeinde gefunden.

### Note

Für jede Großstadt aus `cities_df` (Abschnitt 2) wird zunächst die Großstadt selbst als eigene Zeile mit `size_class="large"` in `matches` aufgenommen (`distance_km=0`, da sie sich nicht "von sich selbst" entfernt befindet) — dadurch enthält `comparison_df` am Ende alle vier Größenklassen als vollständige, gleich aufgebaute Tabelle. Danach wird `get_nearest_comparison_place` (Abschnitt 4) dreimal aufgerufen, einmal je Eintrag im `categories`-Dictionary. Anders als in der Deutschland-Version enthält `categories` kein `place_kind` mehr — da Italien keine Stadtrecht-Unterscheidung kennt, wird jede Kategorie ausschließlich über ihren Populationsbereich definiert:

- **Mittelstadt:** 20.000-149.999 Einwohner
- **Kleinstadt:** 5.000-19.999 Einwohner
- **Gemeinde:** 1-4.999 Einwohner, zusätzlich Mindestabstand `RURAL_MIN_DISTANCE_KM = 15` km, damit kein direkt an die Großstadt angrenzender Vorort als "ländliche Gemeinde" gewählt wird

Wird für eine Kategorie kein Treffer gefunden, wird das ausgegeben und die Kategorie bleibt für diese Großstadt einfach leer, statt den ganzen Lauf abzubrechen. Am Ende wird `matches` in `comparison_df` umgewandelt — das ist die zentrale Tabelle, mit der der Rest des Notebooks arbeitet.

```python
matches = []

RURAL_MIN_DISTANCE_KM = 15

for _, row in cities_df.iterrows():
    large_city = row["place"]
    large_pop = int(row["population"])
    lat = row["lat"]
    lon = row["lon"]

    print(f"Suche Vergleichsorte für {large_city} ...")

    categories = {
        "medium": {
            "min_pop": 20000,
            "max_pop": 149999,
            "min_distance_km": 0
        },
        "small": {
            "min_pop": 5000,
            "max_pop": 19999,
            "min_distance_km": 0
        },
        "rural": {
            "min_pop": 1,
            "max_pop": 4999,
            "min_distance_km": RURAL_MIN_DISTANCE_KM
        }
    }

    matches.append({
        "large_city": large_city,
        "large_city_population": large_pop,
        "comparison_place": large_city,
        "size_class": "large",
        "population": large_pop,
        "distance_km": 0.0,
        "search_radius_km": 0,
        "lat": lat,
        "lon": lon,
        "municipality_key": row.get("municipality_key", None),
        "wikidata_id": row.get("wikidata_id", None)
    })

    for size_class, settings in categories.items():
        result = get_nearest_comparison_place(
            lat=lat,
            lon=lon,
            min_pop=settings["min_pop"],
            max_pop=settings["max_pop"],
            large_city_name=large_city,
            min_distance_km=settings["min_distance_km"]
        )

        if result is not None:
            matches.append({
                "large_city": large_city,
                "large_city_population": large_pop,
                "comparison_place": result["place"],
                "size_class": size_class,
                "population": int(result["population"]),
                "distance_km": float(result["distance_km"]),
                "search_radius_km": int(result["search_radius_km"]),
                "lat": float(result["lat"]),
                "lon": float(result["lon"]),
                "municipality_key": result["municipality_key"],
                "wikidata_id": result["wikidata_id"]
            })
        else:
            print(f"  Kein gültiger {size_class}-Treffer für {large_city} gefunden.")

comparison_df = pd.DataFrame(matches)
comparison_df
```

## 6. Übersichtliche Matching-Tabelle
### Note
`comparison_df` hat eine Zeile pro (Großstadt, Größenklasse)-Kombination — praktisch für die weitere Verarbeitung, aber unübersichtlich zu lesen. Diese Zelle baut daraus eine klassische Übersichtstabelle mit einer Zeile pro Großstadt und je einer Spalte für Mittelstadt/Kleinstadt/Gemeinde (per `pivot`, die Großstadt-Zeilen selbst werden dabei ausgeschlossen, da nur die drei gefundenen Vergleichsorte interessieren). Zeilen sind nach Einwohnerzahl absteigend sortiert, Spalten manuell in der Reihenfolge Mittelstadt/Kleinstadt/Gemeinde (statt der von `pivot` alphabetisch vorgegebenen).

```python
city_order = (
    cities_df
    .sort_values("population", ascending=False)["place"]
    .tolist()
)

comparison_table = comparison_df[
    comparison_df["size_class"].isin(["medium", "small", "rural"])
].pivot(
    index="large_city",
    columns="size_class",
    values="comparison_place"
)

comparison_table = comparison_table[
    ["medium", "small", "rural"]
]

comparison_table = comparison_table.reindex(city_order)

comparison_table = comparison_table.rename(
    columns={
        "medium": "Mittelstadt",
        "small": "Kleinstadt",
        "rural": "Gemeinde"
    }
)

comparison_table.index.name = "Großstadt"

comparison_table
```

## 6b. Karte: ausgewählte Großstädte und ihre Vergleichsorte (Small Multiples)

Alle Großstädte und ihre gefundenen Vergleichsorte (`comparison_df`, inkl. der Großstädte selbst) auf einer Karte Italiens mit Regionen-Grenzen. Die Grenzen kommen aus einer öffentlichen GeoJSON-Datei mit den Umrissen der 20 italienischen Regionen (Projekt "geojson-italy" von openpolis auf GitHub) — abgefragt ganz genauso mit `requests` wie die Wikidata- und Open-Meteo-Anfragen an anderer Stelle im Notebook, also ohne eine zusätzliche Bibliothek wie `geopandas`.

Wie in der Deutschland-Version (siehe dort Abschnitt 8) kann ein Vergleichsort mehreren Großstädten gleichzeitig zugeordnet sein, wenn diese geografisch nah beieinander liegen und sich denselben nächstgelegenen kleineren Ort teilen — gerade in dicht besiedelten Regionen überlagern sich dadurch auf einer einzigen Karte mit vier Farben leicht viele benachbarte Punkte. Die sauberere Lösung: **vier kleine Karten statt einer großen** — eine pro Größenklasse (Small Multiples). Jede Karte zeigt dadurch deutlich weniger Punkte gleichzeitig, wodurch sich auch dicht besiedelte Regionen klar auseinanderhalten lassen; der Vergleich zwischen den Größenklassen bleibt trotzdem möglich, weil alle vier Karten denselben Kartenausschnitt und Maßstab verwenden.

Die Punktgrößen-Kodierung für geteilte Vergleichsorte bleibt zusätzlich erhalten (jetzt mit gemeinsamer Legende unten): Ein Vergleichsort, der mehreren Großstädten zugeordnet ist, erscheint innerhalb seiner eigenen Größenklassen-Karte trotzdem nur als ein einziger, größerer Punkt statt mehrerer sich überlagernder — die beiden Techniken lösen also zwei unterschiedliche Überlagerungsprobleme (zwischen Größenklassen bzw. innerhalb einer Größenklasse).

```python
import matplotlib.pyplot as plt

geojson_url = "https://raw.githubusercontent.com/openpolis/geojson-italy/master/geojson/limits_IT_regions.geojson"
response = requests.get(geojson_url, timeout=30)
response.raise_for_status()
states_geojson = response.json()

# Dieselben Farben wie spaeter bei den Hitzewellen-Grafiken (Abschnitt 16a/16b/16d/16e), damit
# Groessenklassen ueber das ganze Notebook hinweg wiedererkennbar bleiben.
colors = {"large": "steelblue", "medium": "darkorange", "small": "seagreen", "rural": "firebrick"}
class_labels = {"large": "Large city", "medium": "Medium-sized city", "small": "Small town", "rural": "Rural municipality"}
class_order = ["large", "medium", "small", "rural"]

# Pro eindeutiger Koordinate zaehlen, wievielen Großstadtregionen dieser Ort zugeordnet ist --
# ueber eindeutige large_city-Werte, nicht ueber Zeilen, damit eine zufaellige doppelte Zuordnung
# nicht faelschlich mitgezaehlt wuerde.
plot_df = (
    comparison_df
    .groupby(["size_class", "lat", "lon"], observed=True, as_index=False)
    .agg(n_shared=("large_city", "nunique"))
)

fig, axes = plt.subplots(2, 2, figsize=(11, 12), dpi=140, constrained_layout=True)
mean_lat = comparison_df["lat"].mean()

for ax, size_class in zip(axes.flat, class_order):
    # Regionengrenzen zeichnen, bewusst zurueckhaltend (helle Flaeche, duenne Linie), damit
    # die farbigen Orte im Vordergrund stehen. Eine Regionen-Geometrie kann aus mehreren Polygonen
    # bestehen (z.B. Sizilien, Sardinien und kleinere Inseln) -- ueber alle Polygone iterieren.
    for feature in states_geojson["features"]:
        geom = feature["geometry"]
        polygons = geom["coordinates"] if geom["type"] == "MultiPolygon" else [geom["coordinates"]]
        for polygon in polygons:
            exterior = np.asarray(polygon[0])
            ax.fill(exterior[:, 0], exterior[:, 1], facecolor="#f4f5f7", edgecolor="#b5b9c0", linewidth=0.55, zorder=1)

    # Nur eine Größenklasse pro Panel -- das loest das Ueberlagerungsproblem in Ballungsraeumen.
    sub = plot_df[plot_df["size_class"] == size_class]
    ax.scatter(
        sub["lon"], sub["lat"], s=26 * sub["n_shared"], color=colors[size_class],
        alpha=0.78, edgecolor="white", linewidth=0.6, zorder=2
    )
    ax.set_title(f"{class_labels[size_class]} (n = {len(sub)})", fontsize=12, fontweight="bold", color=colors[size_class])
    ax.set_aspect(1 / np.cos(np.deg2rad(mean_lat)))  # Breitengrad-Korrektur, sonst wirkt die Karte zu breit gezogen
    ax.set_xlim(6, 19)
    ax.set_ylim(35, 47.5)
    ax.axis("off")

# Gemeinsame Legende fuer die Punktgroesse -- Referenzwerte 1/2/4/8 plus (falls groesser) der
# tatsaechliche Maximalwert, damit auch der groesste vorkommende Punkt eine exakte Referenz hat.
max_shared = int(plot_df["n_shared"].max())
size_values = sorted(set(v for v in [1, 2, 4, 8, max_shared] if v <= max_shared))
size_handles = [
    plt.scatter([], [], s=26 * value, color="#777777", alpha=0.7, edgecolor="white", label=f"{value}")
    for value in size_values
]
fig.legend(
    handles=size_handles, title="Assigned large-city regions", loc="lower center",
    ncol=len(size_values), frameon=False, bbox_to_anchor=(0.5, -0.015)
)

fig.suptitle("Spatial distribution of study places by size class", fontsize=15, fontweight="bold")
fig.text(
    0.5, -0.035,
    "Point size shows how many large-city regions a place was assigned to as a comparison place. "
    "Map basis: geojson-italy (openpolis).",
    ha="center", fontsize=9, color="#555555"
)

plt.savefig("RQ2_Italy_selected_places_small_multiples.png", dpi=300, bbox_inches="tight", facecolor="white")
plt.show()
```

## 8. Vergleichsorte deduplizieren

`comparison_df` hat eine Zeile pro (Großstadt, Größenklasse)-Kombination (inkl. der Großstadt selbst), aber **nicht entsprechend viele unterschiedliche Orte**. Schaut man sich Abschnitt 6 an, taucht derselbe Vergleichsort häufig für mehrere Großstädte gleichzeitig auf — genau wie in der Deutschland-Version (dort z. B. "Friesenhagen" als Gemeinde-Vergleichsort für gleich mehrere Ruhrgebietsstädte), weil geografisch nah beieinanderliegende Großstädte oft denselben nächstgelegenen kleineren Ort teilen.

Für die Hitzewellen-Berechnung ist das wichtig aus zwei Gründen:

1. **Effizienz:** Für einen physisch selben Ort brauchen wir die Wetterdaten nur einmal zu laden, nicht einmal pro Großstadt, die ihn als Vergleichsort hat.
2. **Korrektheit der späteren Statistik:** Würde man die Hitzewellen eines mehrfach zugeordneten Ortes mehrfach (einmal pro Großstadt, die ihn als Vergleichsort hat) in die Auswertung seiner Größenklasse einfließen lassen, würde dieser eine Ort das Ergebnis der Größenklasse künstlich verzerren (Pseudo-Replikation). Jeder eindeutige Ort soll genau einmal in die Auswertung nach Größenklasse eingehen.

Dedupliziert wird über `municipality_key` (den ISTAT-Gemeindecode) statt über den Namen, aus demselben Grund wie in Abschnitt 1: eindeutiger als der Ortsname.

```python
unique_places_df = (
    comparison_df
    .drop_duplicates(subset="municipality_key")
    .reset_index(drop=True)
)

print(f"comparison_df: {len(comparison_df)} Zeilen")
print(f"unique_places_df: {len(unique_places_df)} eindeutige Orte "
      f"(Ersparnis: {len(comparison_df) - len(unique_places_df)} doppelte Vergleichsorte)")

unique_places_df["size_class"].value_counts().reindex(["large", "medium", "small", "rural"])
```

## 9. Wetter-Client aufbauen

Derselbe Open-Meteo-Client wie in der Deutschland-Version von RQ2: persistenter Cache (Klimadaten der Vergangenheit ändern sich nicht rückwirkend, daher `expire_after=-1`) und eine Retry-Strategie mit steigender Wartezeit bei Rate-Limit-Fehlern (429). `backoff_factor=5`, weil RQ2 (Deutschland wie Italien) mit deutlich mehr eindeutigen Orten viele Batches braucht und dadurch öfter ans Rate-Limit stößt.

**Eigener Cache-Name für Italien:** `CACHE_NAME` heißt hier `openmeteo_cache_italy` statt `openmeteo_cache`, damit die italienischen Wetterdaten in einer eigenen SQLite-Datei landen und die bereits gefüllte `openmeteo_cache.sqlite` der Deutschland-Version unangetastet bleibt — beide Notebooks können dadurch problemlos im selben Ordner liegen.

**Zum Cache-Verhalten:** Die Cache-Datei landet im aktuellen Arbeitsverzeichnis (bei einem normal gestarteten Jupyter-Notebook ist das der Ordner, in dem die `.ipynb`-Datei liegt). Die Zelle gibt beim Start aus, wo genau die Datei liegt und wie viele Einträge schon drin sind, damit sofort sichtbar ist, ob der Cache wiedergefunden wurde — praktisch z.B. wenn du das Notebook mal aus einer anderen IDE oder einem anderen Ordner heraus öffnest, wo der Cache-Name zwar gleich, das Arbeitsverzeichnis aber ein anderes ist.

**Zur Diagnose (jetzt anhand der offiziellen Open-Meteo-Dokumentation bestätigt, siehe Abschnitt 10):** Open-Meteo zählt Anfragen nicht als "1 HTTP-Request = 1 Call", sondern gewichtet nach Orten, Tagen und Variablen. Ein Batch mit 25 Orten über 10 Jahre wiegt deutlich mehr als einer mit 15 Orten — deshalb hat `batch_size=25` das Kontingent schneller aufgebraucht, nicht weil die Anfragerate an sich zu hoch war. Und weil sich das Kontingent über den Tag hinweg nicht erneuert, tauchte der Fehler bei `batch_size=15` immer an fast derselben Stelle auf (nach ~26 Batches) — unabhängig von `pause`. Details und die genaue Formel folgen gleich in Abschnitt 10.

Die Reaktion darauf ändert **nicht** die Anzahl der geladenen Orte (`coords` bleibt exakt gleich lang), sondern nur, wie vorsichtig wir anfragen:

- `batch_size` zurück auf 15 (der nachweislich länger stabile Wert).
- Eine gemeinsame, **eskalierende** Pause (`Throttle`) statt fester `pause`/`cooldown`-Werte: bisher wurde nach einem 429 einmalig 60s gewartet und der Batch gesplittet, aber die Pause zwischen den *folgenden* Batches blieb bei den ursprünglichen 10s — das reicht offenbar nicht, wenn der Server über mehrere Minuten misst. Jetzt wächst die Pause bei jedem 429 dauerhaft (×1.6, gedeckelt bei 180s) und gilt danach für alle weiteren Batches, auch für den Analysezeitraum in Abschnitt 13 (derselbe `Throttle` wird weitergereicht).
- Eine zusätzliche Abkühlpause von 90s direkt vor dem ersten großen Ladevorgang, damit ein eventuell noch "heißes" Rate-Limit-Fenster aus einem vorherigen Versuch sicher abgeklungen ist.

```python
import os
import openmeteo_requests
import requests_cache
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

CACHE_NAME = "openmeteo_cache_italy"  # -> openmeteo_cache_italy.sqlite im aktuellen Arbeitsverzeichnis, eigene Datei getrennt von der Deutschland-Version


def make_client():
    cache_session = requests_cache.CachedSession(
        CACHE_NAME,
        expire_after=-1,
        allowable_codes=[200],
    )
    retry = Retry(
        total=6, backoff_factor=5,
        status_forcelist=[429, 500, 502, 503, 504],
        allowed_methods=["GET"],
    )
    cache_session.mount("https://", HTTPAdapter(max_retries=retry))
    cache_session.headers.update({"User-Agent": "student-project/1.0"})

    n_cached = len(cache_session.cache.responses)
    print(f"Arbeitsverzeichnis: {os.getcwd()}")
    print(f"Cache-Datei: {os.path.join(os.getcwd(), CACHE_NAME)}.sqlite")
    print(f"Bereits gespeicherte Antworten im Cache: {n_cached}")
    if n_cached == 0:
        print("  -> 0 Eintraege: entweder erster Lauf, oder die vorhandene")
        print(f"     {CACHE_NAME}.sqlite mit Fortschritt liegt in einem anderen")
        print("     Ordner als oben angezeigt. Falls ja: Notebook-Kernel neu von")
        print("     dem Ordner aus starten, in dem die Datei tatsaechlich liegt.")

    return openmeteo_requests.Client(session=cache_session)


CLIENT = make_client()
```

## 10. Kernfunktionen: Wetterdaten laden + Hitzewellen erkennen (DWD-Definition)

Die Hitzewellen-Logik basiert auf der offiziellen DWD-Definition, identisch zur Deutschland-Version von RQ2:

- **`fetch_tmax_batch`**: Tages-Maximaltemperatur für mehrere Orte gleichzeitig laden.
- **`fetch_tmax_range`**: teilt einen langen Zeitraum in Jahresblöcke auf, damit einzelne Anfragen klein bleiben.
- **`compute_doy_thresholds`**: pro Kalendertag das 98. Perzentil der Referenzperiode (±15 Tage Toleranz) — "ungewöhnlich heiß" ist ortsspezifisch definiert.
- **`find_dwd_heatwaves`**: mindestens 3 aufeinanderfolgende Tage über dem ortsspezifischen Schwellenwert **und** über der festen 28°C-Grenze (DWD-Definition).
- **`analyze_city_heatwaves_from_data`**: verbindet beide Schritte für einen einzelnen Ort.

Eine zusätzliche Absicherung, die bei einer kleineren Anzahl an Orten nicht nötig wäre: `fetch_tmax_batch` ruft jeden Batch jetzt über `_fetch_tmax_chunk_adaptive` auf. Bricht ein Batch trotz der eingebauten Retry-Logik mit anhaltenden 429-Fehlern ab, wird nicht sofort aufgegeben, sondern eine Abkühlpause eingelegt und der Batch **halbiert** — dieselbe Eskalationsidee wie beim Wikidata-Bulk-Load in Abschnitt 1 (dort Populationsbereich halbieren, hier Batchgröße halbieren), nur diesmal, weil RQ2 mit vielen Orten öfter ans Rate-Limit von Open-Meteo stößt.

Neu (nach dem Fehlschlag mit `batch_size=25`): Pause und Abkühlzeit sind jetzt **eine gemeinsame, eskalierende Größe** (`Throttle`), kein fester Wert mehr. Vorher wurde nach einem 429 zwar der betroffene Batch gesplittet und einmalig gewartet, aber alle *weiteren* Batches liefen sofort wieder mit der ursprünglichen (zu schnellen) Pause weiter — das Objekt hier merkt sich stattdessen die zuletzt nötige Wartezeit und gibt sie an alle folgenden Aufrufe weiter, auch über `fetch_tmax_range`-Aufrufe hinweg (Referenz- und Analysezeitraum teilen sich dasselbe `Throttle`-Objekt).

**Wichtige Korrektur, nachdem der Fehler immer wieder an derselben Stelle auftrat:** Open-Meteo veröffentlicht auf seiner Preis-Seite die tatsächliche Formel, mit der Anfragen gegen das kostenlose Kontingent gezählt werden ([open-meteo.com/en/pricing](https://open-meteo.com/en/pricing)):

`gewicht = anzahl_orte * (anzahl_tage / 14) * (anzahl_variablen / 10)`

Das kostenlose Kontingent ist **600 Anfragen/Minute, 5.000/Stunde, 10.000/Tag** — aber eben *gewichtet*, nicht als reine Anzahl von HTTP-Requests gezählt. Ein einzelner Batch mit 15 Orten über 10 Jahre (≈3650 Tage) und 1 Variable wiegt bereits `15 * (3650/14) * (1/10) ≈ 391` "Anfragen" — nicht 1. Das erklärt, warum der Fehler bisher **immer an fast derselben Stelle** auftrat, egal wie `batch_size` oder `pause` eingestellt waren: Nach ungefähr 26 Batches (zwei Jahrzehnten) ist das **Tageskontingent von 10.000** gewichteten Anfragen aufgebraucht — das ist keine Frage der Anfragegeschwindigkeit, die keine Pause der Welt löst, solange man am selben Tag weiterfragt. Die Zelle unten rechnet das mit den echten Zahlen (`len(coords)`) vor, damit sichtbar wird, wie viele Tage der Lauf real braucht.

Zwei Konsequenzen, beide **ohne** die Anzahl der Orte zu verringern:

1. Referenz- (1961–1990) und Analysezeitraum (1980–2025) überlappen sich in 1980–1990 und wurden bisher **doppelt** geladen. Ab jetzt wird nur noch **ein** durchgehender Zeitraum 1961–2025 geladen und lokal in Referenz- und Analyseteil zerlegt — spart rund 15% des Gewichts, ohne einen einzigen Ort oder Tag wegzulassen.
2. `_fetch_tmax_chunk_adaptive` erkennt jetzt den Unterschied zwischen "Anfragerate kurzzeitig zu hoch" (dagegen hilft Warten und Splitten, wie bisher) und "Kontingent tatsächlich erschöpft" (das erkennt man daran, dass selbst ein einzelner Ort bei maximaler Pause noch scheitert). Im zweiten Fall wird sauber abgebrochen (`QuotaExhausted`) statt sinnlos weiter zu splitten und zu warten — der HTTP-Cache aus Abschnitt 9 ist dauerhaft, also kostet ein erneutes Ausführen derselben Zelle an einem anderen Tag (oder nach Ablauf der Stunde) nur noch die Batches, die wirklich noch fehlen.
3. **Behobener Bug:** Bisher hat `fetch_tmax_batch` nach *jedem* Batch die volle `pause` geschlafen — auch dann, wenn der Batch komplett aus dem Cache kam (also gar keine echte Anfrage an Open-Meteo geschickt wurde). Dadurch sah es bei einem erneuten Lauf so aus, als würde `1961-1970` "wieder von vorne" laufen — in Wirklichkeit kamen die Daten zwar sofort aus dem Cache, aber jeder der 13 Batches hat trotzdem 20+ Sekunden sinnlos geschlafen. Jetzt wird die Dauer jedes Batches gemessen: Unter 2 Sekunden gilt als Cache-Treffer (keine Pause nötig), alles darüber als echte neue Anfrage (Pause wie bisher). Bereits geladene Zeiträume laufen dadurch jetzt in Sekunden statt Minuten durch.

```python
ARCHIVE_URL = "https://archive-api.open-meteo.com/v1/archive"


class Throttle:
    """Merkt sich eine gemeinsame Pausenlänge über den gesamten Ladevorgang hinweg.
    Anders als eine feste Pause WÄCHST diese Pause bei jedem 429 dauerhaft (statt
    danach wieder auf den Ausgangswert zurückzufallen) und gilt für alle folgenden
    Batches -- auch über mehrere fetch_tmax_range-Aufrufe hinweg, wenn dasselbe
    Throttle-Objekt weitergereicht wird."""

    def __init__(self, pause, max_pause=180):
        self.pause = pause
        self.max_pause = max_pause

    def escalate(self):
        self.pause = min(self.pause * 1.6, self.max_pause)
        return self.pause


class QuotaExhausted(Exception):
    """Wird ausgelöst, wenn selbst ein einzelner Ort bei maximaler Pause noch mit
    429 scheitert. Das ist kein Timing-Problem mehr (dagegen würde Warten helfen),
    sondern ein Hinweis, dass das Stunden- oder Tageskontingent von Open-Meteo
    erschöpft ist -- das lässt sich nur durch Abwarten (bis zur nächsten Stunde
    bzw. bis zum nächsten Tag) und erneutes Ausführen derselben Zelle lösen. Dank
    des dauerhaften HTTP-Caches aus Abschnitt 9 kostet das erneute Ausführen nur
    die Batches, die wirklich noch fehlen -- kein Ort geht verloren."""
    pass


def _fetch_tmax_chunk_adaptive(client, chunk, start_date, end_date, throttle, depth=0, max_depth=5):
    """Holt Tmax für einen Batch von Orten. Schlägt der Batch trotz der eingebauten
    Retry-Logik fehl (z.B. anhaltende 429er), erhöht throttle.escalate() dauerhaft
    die Pause für alle folgenden Batches, es wird entsprechend lange gewartet, und
    der Batch selbst wird in der Mitte halbiert -- dieselbe Eskalationsidee wie bei
    _fetch_population_range_adaptive in Abschnitt 1, nur für Batchgröße statt
    Populationsbereich. Wichtig: kein Ort geht dabei verloren, links und rechts
    zusammen ergeben wieder exakt alle Orte des ursprünglichen Batches.

    Scheitert sogar ein einzelner Ort noch bei maximaler Pause, wird nicht weiter
    gesplittet (das würde nur Zeit kosten, ohne etwas zu lösen), sondern
    QuotaExhausted ausgelöst -- das Kontingent ist dann schlicht für den Moment
    aufgebraucht."""
    params = {
        "latitude": [lat for lat, lon in chunk],
        "longitude": [lon for lat, lon in chunk],
        "start_date": start_date, "end_date": end_date,
        "daily": ["temperature_2m_max"],
        "timezone": "Europe/Rome",
    }
    try:
        return client.weather_api(ARCHIVE_URL, params=params)
    except Exception as e:
        if len(chunk) <= 1 and throttle.pause >= throttle.max_pause:
            raise QuotaExhausted(
                f"Auch ein einzelner Ort scheitert noch bei maximaler Pause "
                f"({throttle.pause:.0f}s). Das Kontingent von Open-Meteo ist "
                f"vermutlich für diese Stunde/diesen Tag erschöpft. "
                f"Original-Fehler: {e}"
            ) from e

        if depth >= max_depth:
            raise RuntimeError(
                f"Batch mit {len(chunk)} Orten lässt sich auch nach {depth} Splits "
                f"nicht laden. Original-Fehler: {e}"
            ) from e

        cooldown = throttle.escalate()
        print(f"  ... Batch mit {len(chunk)} Orten fehlgeschlagen ({type(e).__name__}: {e}), "
              f"warte {cooldown:.0f}s und splitte (Pause für alle folgenden Batches jetzt {cooldown:.0f}s)...")
        time.sleep(cooldown)

        mid = len(chunk) // 2
        left = _fetch_tmax_chunk_adaptive(client, chunk[:mid], start_date, end_date, throttle, depth + 1, max_depth)
        right = _fetch_tmax_chunk_adaptive(client, chunk[mid:], start_date, end_date, throttle, depth + 1, max_depth)
        return list(left) + list(right)


def fetch_tmax_batch(client, coords, start_date, end_date, throttle, batch_size=15):
    """Holt Tages-Maximaltemperaturen für mehrere Orte gleichzeitig. Alle Orte aus
    coords werden verarbeitet -- batch_size bestimmt nur, wie sie auf Requests
    aufgeteilt werden, nicht ob welche ausgelassen werden."""
    dfs = []
    n_batches = -(-len(coords) // batch_size)

    for b in range(n_batches):
        i = b * batch_size
        chunk = coords[i:i + batch_size]

        t0 = time.perf_counter()
        responses = _fetch_tmax_chunk_adaptive(client, chunk, start_date, end_date, throttle)
        elapsed = time.perf_counter() - t0

        for response in responses:
            daily = response.Daily()
            tmax = daily.Variables(0).ValuesAsNumpy()
            dates = pd.date_range(
                start=pd.to_datetime(daily.Time(), unit="s", utc=True),
                end=pd.to_datetime(daily.TimeEnd(), unit="s", utc=True),
                freq=pd.Timedelta(seconds=daily.Interval()),
                inclusive="left",
            ).tz_localize(None)
            dfs.append(pd.DataFrame({"date": dates, "tmax": tmax}))

        # Ein Cache-Treffer (Daten schon in openmeteo_cache.sqlite) dauert nur
        # Millisekunden, eine echte neue Anfrage an Open-Meteo dagegen mehrere
        # Sekunden. Nur bei einer ECHTEN Anfrage lohnt sich die Pause -- bei einem
        # Cache-Treffer wurde ja gar keine Anfrage an den Server geschickt, die
        # gebremst werden muesste. Ohne diese Unterscheidung wuerde ein Batch, der
        # komplett aus dem Cache kommt, trotzdem 20+ Sekunden schlafen und ein
        # Zeitraum wie 1961-1970 (13 Batches) faelschlich so wirken, als wuerde er
        # "wieder von vorne" laufen, obwohl in Wirklichkeit gar nichts neu geladen wird.
        cache_hit = elapsed < 2.0
        status = "Cache-Treffer" if cache_hit else f"{elapsed:.1f}s, neu geladen"
        print(f"  Batch {b+1}/{n_batches} ok ({len(chunk)} Orte, {status}, aktuelle Pause: {throttle.pause:.0f}s)")
        if not cache_hit:
            time.sleep(throttle.pause)

    return dfs


def fetch_tmax_range(client, coords, start_date, end_date, throttle, batch_size=15, year_chunk=10):
    """Holt Tmax über einen langen Zeitraum, aufgeteilt in Orts-Batches UND Jahres-Blöcke.
    coords wird unverändert und vollständig durchgereicht -- die Anzahl der Orte
    ändert sich durch batch_size/year_chunk nie, nur die Anzahl der HTTP-Requests."""
    start_year, end_year = int(start_date[:4]), int(end_date[:4])

    year_ranges = []
    y = start_year
    while y <= end_year:
        y2 = min(y + year_chunk - 1, end_year)
        year_ranges.append((f"{y}-01-01", f"{y2}-12-31"))
        y = y2 + 1

    all_parts = {i: [] for i in range(len(coords))}
    for chunk_start, chunk_end in year_ranges:
        print(f"Zeitraum {chunk_start[:4]}-{chunk_end[:4]}:")
        t0 = time.perf_counter()
        dfs = fetch_tmax_batch(client, coords, chunk_start, chunk_end, throttle, batch_size=batch_size)
        for i, df in enumerate(dfs):
            all_parts[i].append(df)
        if time.perf_counter() - t0 > 2.0:  # nur pausieren, wenn im Zeitraum echt neu geladen wurde
            time.sleep(throttle.pause)

    return [pd.concat(all_parts[i], ignore_index=True).sort_values("date").reset_index(drop=True)
            for i in range(len(coords))]


THROTTLE = Throttle(pause=20)


def compute_doy_thresholds(ref_df, window=15, percentile=98):
    """Schwellenwert pro Kalendertag: 98. Perzentil aus ±15 Tagen der Referenzperiode."""
    doy = ref_df["doy"].to_numpy()
    tmax = ref_df["tmax"].to_numpy()
    thresholds = {}

    for d in range(1, 366):
        dist = np.abs(doy - d)
        dist = np.minimum(dist, 365 - dist)
        thresholds[d] = np.percentile(tmax[dist <= window], percentile)

    return thresholds


def find_dwd_heatwaves(df, thresholds, min_days=3, fixed_threshold=28.0):
    """Hitzewelle = mind. 3 Tage in Folge über Klima-Schwelle UND über 28°C."""
    df = df.copy()
    df["doy"] = df["date"].dt.dayofyear.clip(upper=365)
    df["threshold"] = df["doy"].map(thresholds)
    df["is_hot"] = (df["tmax"] > df["threshold"]) & (df["tmax"] > fixed_threshold)

    heatwaves, start = [], None
    for i, row in df.iterrows():
        if row["is_hot"] and start is None:
            start = i
        elif not row["is_hot"] and start is not None:
            segment = df.loc[start:i - 1]
            if len(segment) >= min_days:
                heatwaves.append({
                    "start": segment["date"].iloc[0].date(),
                    "end": segment["date"].iloc[-1].date(),
                    "duration_days": len(segment),
                    "max_temp": round(segment["tmax"].max(), 1),
                    "avg_temp": round(segment["tmax"].mean(), 1)
                })
            start = None

    if start is not None:
        segment = df.loc[start:]
        if len(segment) >= min_days:
            heatwaves.append({
                "start": segment["date"].iloc[0].date(),
                "end": segment["date"].iloc[-1].date(),
                "duration_days": len(segment),
                "max_temp": round(segment["tmax"].max(), 1),
                "avg_temp": round(segment["tmax"].mean(), 1)
            })

    return heatwaves


def analyze_city_heatwaves_from_data(ref_df, analysis_df):
    """Verbindet Schwellenwert-Berechnung und Hitzewellen-Suche für einen Ort."""
    ref_df = ref_df.copy()
    ref_df["doy"] = ref_df["date"].dt.dayofyear.clip(upper=365)
    thresholds = compute_doy_thresholds(ref_df)
    return find_dwd_heatwaves(analysis_df, thresholds)
```

## 12. Wie viele Tage braucht der volle Lauf realistisch?

Referenz- und Analysezeitraum werden ab jetzt als **ein** durchgehender Zeitraum 1961–2025 geladen (statt zwei sich überlappender), um die doppelte Abfrage von 1980–1990 zu vermeiden. Mit der Gewichtsformel von Open-Meteo lässt sich vorher ausrechnen, wie viele Tage das bei einem Tageskontingent von 10.000 realistisch dauert — mit den echten Zahlen aus `coords`, nicht geschätzt.

```python
FULL_START, FULL_END = "1961-01-01", "2025-12-31"
n_days_full = (pd.Timestamp(FULL_END) - pd.Timestamp(FULL_START)).days + 1
n_places = len(coords)
weight_total = n_places * (n_days_full / 14) * (1 / 10)  # 1 Variable: temperature_2m_max

daily_quota = 10_000
print(f"Eindeutige Orte: {n_places}")
print(f"Durchgehender Zeitraum {FULL_START} bis {FULL_END}: {n_days_full} Tage")
print(f"Geschätztes Gewicht laut Open-Meteo-Formel (Orte * Tage/14 * Variablen/10): {weight_total:,.0f}")
print(f"Freies Tageskontingent: {daily_quota:,}")
print(f"-> reicht für ca. {weight_total / daily_quota:.1f} Tage Laufzeit (bei taeglich neu verfuegbarem Kontingent)")
```

Das Ergebnis zeigt: Bei knapp 200 eindeutigen Orten und 65 Jahren übersteigt das benötigte Gewicht das Tageskontingent von 10.000 üblicherweise um ein Mehrfaches — unabhängig von `batch_size` oder `pause` (die genaue Anzahl der Orte für Italien ergibt sich erst aus dem tatsächlichen Lauf von Abschnitt 5, dürfte aber in einer ähnlichen Größenordnung wie in der Deutschland-Version liegen). Das ist also kein Bug, sondern eine harte Grenze des kostenlosen Open-Meteo-Kontingents. Die Zelle unten läuft deshalb absichtlich **nicht** in einem Rutsch durch: Sobald `QuotaExhausted` auftritt, bricht sie sauber ab, statt sinnlos weiterzusplitten. Der HTTP-Cache aus Abschnitt 9 ist dauerhaft (`expire_after=-1`), also einfach **dieselbe Zelle später erneut ausführen** (nach ein paar Stunden bzw. am nächsten Tag) — bereits geladene Orte/Zeiträume kommen sofort aus dem Cache, es wird nur an der Stelle weitergemacht, an der zuletzt abgebrochen wurde. Kein Ort wird dabei übersprungen oder reduziert. Bei einem frischen Lauf (leerer `openmeteo_cache_italy.sqlite`) ist deshalb damit zu rechnen, dass diese Zelle mehrfach unterbrochen wird und sich der volle Ladevorgang über mehrere Tage zieht, bis der komplette Zeitraum für alle Orte im Cache liegt — danach läuft dieselbe Zelle in Sekunden durch, weil dann jeder Batch ein Cache-Treffer ist.

## 13. Referenz- und Analysezeitraum laden (1961–2025, durchgehend)

Ein einziger durchgehender Ladevorgang für 1961–2025. Danach wird lokal (kostenlos, ohne weitere Anfragen) in Referenzperiode (1961–1990) und Analysezeitraum (1980–2025) aufgeteilt.

**Automatischer Retry statt manuellem Neustart:** Weil das Tageskontingent (siehe Abschnitt 12) für die volle Datenmenge nicht in einem Tag reicht, läuft diese Zelle in einer Schleife: Tritt `QuotaExhausted` auf, wartet sie automatisch `RETRY_WAIT_HOURS` Stunden und versucht es dann erneut — ohne dass du die Zelle von Hand neu ausführen musst. Dank des dauerhaften Caches beginnt jeder Versuch nicht bei null, sondern genau dort, wo der vorherige aufgehört hat.

Bei einem frischen Lauf ist normal, dass diese Zelle beim ersten Ausführen einige Zeit braucht und ggf. mehrfach über `QuotaExhausted` automatisch wartet und erneut versucht, bis alle Orte für den gesamten Zeitraum geladen sind. Sobald alle Orte vollständig im Cache liegen, läuft dieselbe Zelle bei einem erneuten Ausführen nur noch wenige Sekunden (jeder Batch ist dann ein "Cache-Treffer", die `while`-Schleife oben wird praktisch nicht mehr gebraucht). Der Code bleibt in beiden Fällen unverändert derselbe — inklusive Batch-Aufteilung, adaptivem Splitten bei 429-Fehlern und automatischem Retry über das Tageskontingent hinweg.

```python
import time as _time  # nur fuer die Zeitstempel in den Log-Ausgaben unten

RETRY_WAIT_HOURS = 2

# Abkühlpause vor dem ersten großen Ladevorgang: falls ein vorheriger Lauf bereits
# mehrfach an 429ern gescheitert ist, war das Rate-Limit-Fenster von Open-Meteo beim
# Neustart evtl. noch nicht zurückgesetzt. Kostet nur einmalig 90s, verhindert aber,
# dass wir sofort wieder in dasselbe Limit laufen.
time.sleep(90)

while True:
    try:
        full_dfs = fetch_tmax_range(CLIENT, coords, FULL_START, FULL_END, THROTTLE, batch_size=15, year_chunk=10)
        break  # alles komplett geladen -- Schleife verlassen
    except QuotaExhausted as e:
        now = _time.strftime("%Y-%m-%d %H:%M:%S")
        print(f"[{now}] Kontingent erschöpft: {e}")
        print(f"[{now}] Bereits geladene Orte/Zeiträume liegen sicher im Cache (openmeteo_cache).")
        print(f"[{now}] Warte {RETRY_WAIT_HOURS} Stunden und versuche es automatisch erneut...")
        # Nach der langen Wartezeit ist ein kurzzeitiges Rate-Limit-Fenster sicher
        # abgeklungen -- die Pause wieder auf den Ausgangswert zuruecksetzen, statt
        # bei der zuletzt eskalierten (evtl. sehr hohen) Pause weiterzumachen.
        THROTTLE.pause = 20
        time.sleep(RETRY_WAIT_HOURS * 60 * 60)

print(f"Fertig! Alle {len(coords)} Orte, Zeitraum {FULL_START} bis {FULL_END}, vollstaendig geladen.")

ref_dfs = [df[(df["date"] >= FULL_START) & (df["date"] <= "1990-12-31")].reset_index(drop=True)
           for df in full_dfs]
analysis_dfs = [df[(df["date"] >= "1980-01-01") & (df["date"] <= FULL_END)].reset_index(drop=True)
                for df in full_dfs]
```

## 14. Hitzewellen pro eindeutigem Vergleichsort berechnen

Für jeden Ort in `unique_places_df` wird `analyze_city_heatwaves_from_data` aufgerufen. Jede gefundene Hitzewelle landet als eigene Zeile in `results_df` (nicht jeder Ort — ein Ort mit z. B. 5 Hitzewellen erzeugt 5 Zeilen), zusammen mit `municipality_key` und `size_class`, damit die spätere Aggregation (Abschnitt 15) weiß, zu welcher Größenklasse und welchem physischen Ort sie gehört. `idx` läuft dabei synchron zu `ref_dfs`/`analysis_dfs` aus Abschnitt 13 mit, da beide Listen über dieselbe Reihenfolge wie `coords`/`unique_places_df` aufgebaut wurden. Schlägt die Berechnung für einen einzelnen Ort fehl, bricht das nicht den ganzen Lauf ab — der Ort wird übersprungen und am Ende in `failed_places` sichtbar aufgelistet.

```python
results = []
failed_places = []

for idx, row in unique_places_df.iterrows():
    try:
        heatwaves = analyze_city_heatwaves_from_data(ref_dfs[idx], analysis_dfs[idx])
        for hw in heatwaves:
            results.append({
                "comparison_place": row["comparison_place"],
                "size_class": row["size_class"],
                "municipality_key": row["municipality_key"],
                "population": row["population"],
                **hw
            })
    except Exception as e:
        print(f"  Übersprungen wegen Fehler ({row['comparison_place']}): {e}")
        failed_places.append(row["comparison_place"])

results_df = pd.DataFrame(results)
results_df["year"] = pd.to_datetime(results_df["start"]).dt.year

print(results_df)
print(f"\nGefundene Hitzewellen gesamt: {len(results_df)}")
print(f"Fehlgeschlagen: {failed_places}")
```

## 15. Kennzahlen pro Größenklasse aggregieren

Wichtig: aggregiert wird auf `unique_places_df`/`results_df` (eindeutige Orte), **nicht** auf dem vollen, nicht-deduplizierten `comparison_df` — sonst würde ein mehrfach verwendeter Vergleichsort mehrfach in die Statistik einer Größenklasse eingehen (siehe Abschnitt 8).

`place_level` fasst pro eindeutigem Ort zusammen: wie viele Hitzewellen er im Analysezeitraum hatte, und die mittlere Intensität/Dauer dieser Hitzewellen. Orte ganz ohne Hitzewelle bekommen `n_heatwaves = 0`, aber `NaN` bei Temperatur/Dauer (ein "Durchschnitt" über null Ereignisse ergibt keinen sinnvollen Wert).

`size_order` legt die Größenklassen als geordnete Kategorie fest (large < medium < small < rural), damit sie in Tabellen und Diagrammen im ganzen Notebook konsistent in dieser Reihenfolge erscheinen — passend zur Formulierung der Forschungsfrage "from large cities to rural municipalities". `summary` aggregiert `place_level` anschließend ein weiteres Mal, von "ein Wert pro Ort" zu "ein Wert pro Größenklasse" — das ist die Kennzahlen-Tabelle, aus der die Balkendiagramme in Abschnitt 16 entstehen.

```python
size_order = pd.CategoricalDtype(["large", "medium", "small", "rural"], ordered=True)

ANALYSIS_YEARS = 2025 - 1980 + 1

place_level = unique_places_df[["municipality_key", "size_class", "comparison_place"]].merge(
    results_df.groupby("municipality_key").agg(
        n_heatwaves=("start", "count"),
        mean_max_temp=("max_temp", "mean"),
        mean_avg_temp=("avg_temp", "mean"),
        mean_duration=("duration_days", "mean"),
    ),
    on="municipality_key", how="left"
)
place_level["n_heatwaves"] = place_level["n_heatwaves"].fillna(0).astype(int)
place_level["size_class"] = place_level["size_class"].astype(size_order)

summary = place_level.groupby("size_class", observed=False).agg(
    n_orte=("municipality_key", "nunique"),
    gesamt_hitzewellen=("n_heatwaves", "sum"),
    mean_hitzewellen_pro_ort=("n_heatwaves", "mean"),
    mean_max_temp=("mean_max_temp", "mean"),
    mean_avg_temp=("mean_avg_temp", "mean"),
    mean_duration=("mean_duration", "mean"),
).reindex(["large", "medium", "small", "rural"])

summary["hitzewellen_pro_ort_und_jahr"] = summary["mean_hitzewellen_pro_ort"] / ANALYSIS_YEARS
summary = summary.round(3)

summary
```

## 16a. Regionale Differenz zur Großstadt (Hauptgrafik)

Die Kennzahlen-Tabelle `summary` aus Abschnitt 15 vergleicht die Größenklassen bisher nur absolut — aber die Forschungsfrage fragt explizit nach Unterschieden **"within the same region"**. Diese Grafik bildet das direkt ab: Für jede Großstadt-Region wird die Differenz zwischen dem Vergleichsort und der jeweils eigenen Großstadt berechnet (`Δ = Wert Vergleichsort − Wert Großstadt`), statt Größenklassen unabhängig voneinander zu betrachten. Das kontrolliert dafür, dass manche Regionen grundsätzlich wärmer sind als andere — ein Effekt, der sonst mit dem eigentlich interessierenden Stadtgrößen-Effekt vermischt würde.

Jedes Panel zeigt jetzt alle vier Größenklassen: **Großstadt (Referenz) – Mittelstadt – Kleinstadt – Gemeinde**. Großstadt steht fest bei Δ = 0 (schwarzer Punkt, bewusst ohne Konfidenzintervall — Großstadt minus Großstadt ist immer exakt 0). Für Mittelstadt, Kleinstadt und Gemeinde zeigt der schwarze Punkt den Mittelwert der regionalen Differenzen, der Fehlerbalken das 95%-Bootstrap-Konfidenzintervall dieses Mittelwerts (`bootstrap_ci`, dieselbe Bootstrap-Methode wie bisher über `seaborn.algorithms.bootstrap`, nur direkt aufgerufen statt über `sns.pointplot`). Die gestrichelte Nulllinie bleibt als Referenz erhalten, die Bildunterschrift fasst die Legende in Textform zusammen.

Ein Detail zur Berechnungsgrundlage (unverändert): `comparison_df` (nicht `unique_places_df`) wird hier bewusst **ohne** Deduplizierung verwendet — ein Vergleichsort, der sich mehrere Großstädte teilt (Abschnitt 8), liefert dadurch mehrere Δ-Werte, einen pro Region-Paarung. Das ist hier korrekt, weil jede Paarung einen eigenen Vergleich zu einer jeweils anderen Großstadt darstellt.

**Abhängigkeit:** Diese Grafik nutzt `seaborn` zusätzlich zu `matplotlib` (für `sns.algorithms.bootstrap` in `bootstrap_ci`) — falls das noch nicht installiert ist: `pipenv install seaborn`.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

class_order = ["large", "medium", "small", "rural"]
class_labels = {"large": "Large city\n(reference)", "medium": "Medium-sized city", "small": "Small town", "rural": "Rural municipality"}
x_positions = {size_class: i for i, size_class in enumerate(class_order)}

regional = comparison_df[["large_city", "municipality_key", "size_class"]].merge(
    place_level.drop(columns="size_class"),
    on="municipality_key",
    how="left"
)
regional["size_class"] = pd.Categorical(regional["size_class"], categories=class_order, ordered=True)
regional["heatwaves_per_year"] = regional["n_heatwaves"] / ANALYSIS_YEARS

metrics = [
    ("heatwaves_per_year", "Frequency", "Difference in heatwaves per place and year"),
    ("mean_max_temp", "Peak temperature of heatwaves", "Temperature difference (°C)"),
    ("mean_avg_temp", "Average temperature of heatwaves", "Temperature difference (°C)"),
    ("mean_duration", "Duration", "Difference in days"),
]

# Fuer jede Kennzahl: Differenz zur jeweils eigenen Großstadt berechnen. Die "large"-Zeilen
# selbst werden dabei auf sich selbst gemappt und ergeben deshalb immer exakt 0.
for column, _, _ in metrics:
    large_city_values = regional[regional["size_class"] == "large"].set_index("large_city")[column]
    regional[f"delta_{column}"] = regional[column] - regional["large_city"].map(large_city_values)


def bootstrap_ci(values, n_boot=1000, confidence=95):
    """Mittelwert + Bootstrap-Konfidenzintervall einer Wertereihe -- dieselbe Bootstrap-Methode,
    die bisher intern in sns.pointplot steckte, hier nur direkt aufgerufen. Dadurch bleibt volle
    Kontrolle darueber, dass fuer "large" unten bewusst KEIN Intervall berechnet/gezeichnet wird."""
    values = np.asarray(values)
    values = values[~np.isnan(values)]
    if len(values) == 0:
        return np.nan, np.nan, np.nan
    boot_means = sns.algorithms.bootstrap(values, func=np.mean, n_boot=n_boot, seed=42)
    alpha = (100 - confidence) / 2
    return values.mean(), np.percentile(boot_means, alpha), np.percentile(boot_means, 100 - alpha)


fig, axes = plt.subplots(2, 2, figsize=(13, 9))
for ax, (column, title, ylabel) in zip(axes.flat, metrics):
    # Großstadt-Referenzpunkt bei (0, 0): kein Bootstrap-Fehlerbalken, weil Großstadt minus
    # Großstadt per Definition immer exakt 0 ist -- ein Konfidenzintervall dafuer waere nur
    # vorgetaeuschte Praezision.
    ax.scatter(x_positions["large"], 0, color="black", s=55, zorder=5)

    for size_class in ["medium", "small", "rural"]:
        mean_value, ci_low, ci_high = bootstrap_ci(regional.loc[regional["size_class"] == size_class, f"delta_{column}"])
        yerr = [[mean_value - ci_low], [ci_high - mean_value]]
        ax.errorbar(
            x_positions[size_class], mean_value, yerr=yerr,
            fmt="o", color="black", markersize=7, capsize=7, capthick=2, linewidth=2, zorder=4
        )

    ax.axhline(0, color="gray", linestyle="--", linewidth=1)
    ax.set_xticks(range(len(class_order)))
    ax.set_xticklabels([class_labels[c] for c in class_order])
    ax.set_title(title)
    ax.set_xlabel("")
    ax.set_ylabel(ylabel)
    ax.grid(axis="y", alpha=0.2)

fig.suptitle("Differences between place sizes within the same region", fontsize=15)
fig.text(
    0.5, 0.01,
    "Δ = comparison place − corresponding large city within the same region. "
    "Black dot = mean, error bar = 95% confidence interval.",
    ha="center", fontsize=10
)
plt.tight_layout(rect=[0, 0.04, 1, 0.96])
plt.savefig("RQ2_Italy_regional_delta.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 16b. Zeitlicher Verlauf: Entwicklung 1980–2025 je Größenklasse

Die Balken oben zeigen nur den Durchschnitt über den *gesamten* Analysezeitraum. Hier zusätzlich der Verlauf über die Jahre, nur eben nicht über alle Städte gepoolt, sondern getrennt nach Größenklasse: pro Jahr und Größenklasse werden Häufigkeit (Hitzewellen pro Ort), Intensität (mittlere Spitzen- **und** Durchschnittstemperatur, passend zur Formulierung der Forschungsfrage) und Dauer berechnet.

Jede Größenklasse wird als gleitender 5-Jahres-Durchschnitt (`rolling(5, center=True)`) dargestellt, damit der langfristige Trend erkennbar bleibt statt in den Jahr-zu-Jahr-Schwankungen unterzugehen. Auf zusätzliche rohe Jahreswerte im Hintergrund wird bei allen vier Kennzahlen verzichtet — in den italienischen Daten hatten praktisch alle Jahre und Größenklassen mindestens eine Hitzewelle, sodass sich die vier Rohdaten-Linien ohnehin nur überlagert hätten, ohne zusätzliche Information zu liefern. Die reinen Jahreswerte bleiben aber für alle vier Kennzahlen weiterhin in `yearly` verfügbar, z. B. für die Trendstatistik in Abschnitt 16c.

Aus demselben Grund gibt es hier auch kein Zoom-Fenster wie in einigen anderen Diagrammen: Ein Zoom auf Jahre, in denen die Großstadt Hitzewellen hatte, während Mittelstadt, Kleinstadt und Gemeinde bei 0 lagen, wäre für Italien nicht aussagekräftig, da ein solcher Kontrast in den Daten kaum vorkommt.

Jede der vier Kennzahlen bekommt ihr eigenes, vollformatiges Diagramm (statt eines gemeinsamen 2×2-Rasters) — dadurch hat jede einzelne Kennzahl mehr Platz, was bei vier sich überlagernden Linien pro Diagramm die Lesbarkeit verbessert. Die X-Achse zeigt jedes einzelne Jahr als eigenen Tick (statt der matplotlib-Standardschritte von 10 Jahren), damit sich jeder Jahrgang direkt ablesen lässt.

Als Nenner für "Hitzewellen pro Ort" zählt hier **nicht** einfach die Gesamtzahl aller eindeutigen Orte, sondern pro Jahr nur die Orte, für die in diesem Jahr überhaupt Temperaturdaten vorliegen (`n_orte_mit_daten`) — bei vollständigem Cache sind das praktisch immer alle Orte, die Prüfung bleibt aber als Absicherung stehen. Jahre, in denen dennoch weniger als 90 % der Orte einer Größenklasse Daten haben, werden unten explizit aufgelistet.

```python
n_orte_per_class = place_level.groupby("size_class", observed=False)["municipality_key"].nunique()

# Fuer jeden Ort: in welchen Jahren gibt es ueberhaupt Temperaturdaten? (unabhaengig davon,
# ob in diesem Jahr eine Hitzewelle gefunden wurde -- 0 Hitzewellen ist ein gueltiger Wert,
# fehlende Daten sind etwas anderes). Das ergibt den korrekten, jahresabhaengigen Nenner.
place_years = []
for idx, row in unique_places_df.iterrows():
    for y in analysis_dfs[idx]["date"].dt.year.unique():
        place_years.append({"year": int(y), "size_class": row["size_class"]})
coverage = (pd.DataFrame(place_years)
            .groupby(["year", "size_class"], observed=False).size()
            .rename("n_orte_mit_daten").reset_index())

all_years = pd.DataFrame(
    [(y, s) for y in range(1980, 2026) for s in ["large", "medium", "small", "rural"]],
    columns=["year", "size_class"]
)

yearly = results_df.groupby(["year", "size_class"], observed=False).agg(
    n_heatwaves=("start", "count"),
    mean_max_temp=("max_temp", "mean"),
    mean_avg_temp=("avg_temp", "mean"),
    mean_duration=("duration_days", "mean"),
).reset_index()

yearly = all_years.merge(yearly, on=["year", "size_class"], how="left").merge(
    coverage, on=["year", "size_class"], how="left"
)
yearly["n_heatwaves"] = yearly["n_heatwaves"].fillna(0)
yearly["n_orte_mit_daten"] = yearly["n_orte_mit_daten"].fillna(0).astype(int)
yearly["hitzewellen_pro_ort"] = np.where(
    yearly["n_orte_mit_daten"] > 0, yearly["n_heatwaves"] / yearly["n_orte_mit_daten"], np.nan
)
yearly["size_class"] = yearly["size_class"].astype(size_order)

colors = {"large": "steelblue", "medium": "darkorange", "small": "seagreen", "rural": "firebrick"}
class_labels = {"large": "Large city (≥150,000)", "medium": "Medium-sized city (20,000–149,999)", "small": "Small town (5,000–19,999)", "rural": "Rural municipality (1–4,999)"}

# Vier eigenstaendige, vollformatige Diagramme statt eines gemeinsamen 2x2-Rasters --
# jede Kennzahl bekommt dadurch mehr Platz fuer die vier ueberlagerten Groessenklassen-Linien.
# Vierter Eintrag pro Kennzahl: Beschriftung der Y-Achse, damit sofort klar ist, in welcher
# Einheit die Werte gemessen werden (bisher hatte die Y-Achse gar keine Beschriftung).
metrics_yearly = [
    (
        "hitzewellen_pro_ort",
        "Frequency of heatwaves by city size",
        "Average heatwaves per place per year",
        "RQ2_Italy_heatwaves_yearly_trend_frequency.png"
    ),
    (
        "mean_max_temp",
        "Average peak temperature of heatwaves by city size",
        "Average peak temperature (°C)",
        "RQ2_Italy_heatwaves_yearly_trend_peak_temp.png"
    ),
    (
        "mean_avg_temp",
        "Average heatwave temperature by city size",
        "Average heatwave temperature (°C)",
        "RQ2_Italy_heatwaves_yearly_trend_avg_temp.png"
    ),
    (
        "mean_duration",
        "Average duration of heatwaves by city size",
        "Average heatwave duration (days)",
        "RQ2_Italy_heatwaves_yearly_trend_duration.png"
    ),
]

for col, title, ylabel, filename in metrics_yearly:
    fig, ax = plt.subplots(figsize=(12, 6))
    for size_class in ["large", "medium", "small", "rural"]:
        sub = yearly[yearly["size_class"] == size_class].sort_values("year")

        # Geglätteter 5-Jahres-Trend: zentriertes rollierendes Fenster (Jahr selbst + je 2 Jahre
        # davor/danach). min_periods=1 sorgt dafuer, dass auch am Rand der Zeitreihe (1980/2025),
        # wo kein volles 5-Jahres-Fenster existiert, trotzdem ein Wert berechnet wird statt NaN.
        trend = sub[col].rolling(5, min_periods=1, center=True).mean()

        # Nur die geglaettete Trendlinie wird gezeigt -- auf zusaetzliche rohe Jahreswerte im
        # Hintergrund wird bei allen vier Kennzahlen verzichtet. In den italienischen Daten
        # hatten praktisch alle Jahre und Groessenklassen mindestens eine Hitzewelle, sodass
        # sich vier Rohdaten-Linien ohnehin nur ueberlagert und kaum zusaetzliche Information
        # geliefert haetten.
        ax.plot(sub["year"], trend, color=colors[size_class], linewidth=2.5, label=class_labels[size_class])

    ax.set_title(title, fontsize=14)
    ax.set_xlabel("Year")
    ax.set_ylabel(ylabel)
    # Jedes einzelne Jahr als Tick zeigen (nicht nur in 10er-Bloecken wie der matplotlib-Standard),
    # damit sich einzelne Jahre auf der Achse direkt ablesen lassen.
    ax.set_xticks(range(1980, 2026, 5))
    ax.tick_params(axis="x", rotation=90, labelsize=8)
    ax.grid(alpha=0.3)
    ax.legend(loc="upper left", ncol=2)

    plt.tight_layout()
    plt.savefig(filename, dpi=150)
    plt.show()

# Jahre mit lueckenhafter Abdeckung (<90% der Orte einer Groessenklasse) auflisten, damit
# klar ist, welche Linienabschnitte noch auf wenigen Orten beruhen.
n_orte_erwartet = yearly["size_class"].astype(str).map(n_orte_per_class.to_dict())
thin_years = yearly[yearly["n_orte_mit_daten"] < 0.9 * n_orte_erwartet]
if len(thin_years):
    print("Jahre mit <90% Orte-Abdeckung je Größenklasse (Linie dort nur eingeschränkt aussagekräftig):")
    for size_class, grp in thin_years.groupby("size_class", observed=False):
        years_list = sorted(grp["year"].tolist())
        if years_list:
            print(f"  {class_labels[size_class]}: {years_list[0]}–{years_list[-1]} ({len(years_list)} Jahre)")
else:
    print("Alle Jahre haben in jeder Größenklasse ≥90% Orte-Abdeckung.")
```

## 16c. Trendstatistik je Größenklasse (lineare Regression)

Ein Diagramm zeigt einen Trend nur visuell — hier wird er zusätzlich in Zahlen gefasst: pro Größenklasse und Kennzahl eine lineare Regression über die Jahre (`scipy.stats.linregress`). **slope** ist die Änderung pro Jahr, der **p-Wert** sagt, wie unwahrscheinlich ein mindestens so starker Trend rein zufällig wäre (Faustregel: signifikant ab p < 0,05), **R²** zeigt, wie viel der Jahr-zu-Jahr-Schwankung sich allein durch die Zeit erklären lässt. `n` ist die Anzahl der Jahre, die tatsächlich in die Rechnung eingegangen sind (bei den Temperatur-/Dauerspalten kleiner als 46, weil Jahre ganz ohne Hitzewelle dort als NaN behandelt und rausgefiltert werden).

Es gibt hier vier Gruppen gleichzeitig — deshalb als eine Tabelle statt einzelner Print-Zeilen, damit sich die vier Größenklassen pro Kennzahl direkt vergleichen lassen.

```python
from scipy import stats

def trend_table(yearly_subset):
    rows = []
    for size_class in ["large", "medium", "small", "rural"]:
        sub = yearly_subset[yearly_subset["size_class"] == size_class]
        for col, label in [("hitzewellen_pro_ort", "Häufigkeit"), ("mean_max_temp", "Spitzentemp."),
                            ("mean_avg_temp", "Durchschnittstemp."), ("mean_duration", "Dauer")]:
            valid = sub.dropna(subset=[col])
            if len(valid) < 2:
                rows.append({"size_class": class_labels[size_class], "kennzahl": label,
                              "slope_pro_jahr": np.nan, "p_wert": np.nan, "r2": np.nan,
                              "signifikant": "zu wenig Daten", "n": len(valid)})
                continue
            slope, intercept, r, p, se = stats.linregress(valid["year"], valid[col])
            rows.append({"size_class": class_labels[size_class], "kennzahl": label,
                         "slope_pro_jahr": round(slope, 4), "p_wert": round(p, 4),
                         "r2": round(r**2, 3), "signifikant": "ja" if p < 0.05 else "nein",
                         "n": len(valid)})
    return pd.DataFrame(rows)

print("Trendstatistik, gesamter Zeitraum (1980–2025):")
trend_full = trend_table(yearly)
trend_full
```

Referenzperiode (1961–1990) und Analysezeitraum (1980–2025) überlappen sich in den Jahren 1980–1990. Um auszuschließen, dass diese Überlappung die Trendberechnung verzerrt, hier dieselbe Regression noch einmal nur für 1991–2025.

```python
yearly_no_overlap = yearly[yearly["year"] >= 1991]
trend_no_overlap = trend_table(yearly_no_overlap)
trend_no_overlap
```

## 17. Statistischer Test zwischen den Größenklassen

Pro Kennzahl ein Kruskal-Wallis-Test (nichtparametrisches Gegenstück zur einfaktoriellen ANOVA — braucht keine Normalverteilung, was bei Hitzewellen-Zähldaten und -Temperaturen die robustere Annahme ist) über die vier Größenklassen. Getestet wird auf `place_level`, also ein Wert pro eindeutigem Ort — nicht pro Einzelereignis —, damit ein Ort mit vielen Hitzewellen das Ergebnis nicht überproportional durch Pseudo-Replikation dominiert. Je Kennzahl werden die vier Größenklassen-Gruppen gebildet, `NaN`-Werte (Orte ohne jede Hitzewelle) und leere Gruppen vorher entfernt, und mit `scipy.stats.kruskal` verglichen. `p < 0.05` gilt dabei als übliche Signifikanzschwelle.

```python
from scipy import stats

metrics = [
    ("n_heatwaves", "Häufigkeit (Hitzewellen pro Ort)"),
    ("mean_max_temp", "Intensität (Spitzentemperatur)"),
    ("mean_avg_temp", "Intensität (Durchschnittstemperatur)"),
    ("mean_duration", "Dauer"),
]

for col, label in metrics:
    groups = [
        place_level.loc[place_level["size_class"] == cls, col].dropna()
        for cls in ["large", "medium", "small", "rural"]
    ]
    groups = [g for g in groups if len(g) > 0]

    if len(groups) < 2:
        print(f"{label}: zu wenige Gruppen mit Daten für einen Test.")
        continue

    h, p = stats.kruskal(*groups)
    signif = "signifikant" if p < 0.05 else "nicht signifikant"
    print(f"{label}: H={h:.3f}, p={p:.4f} ({signif}), n pro Gruppe = {[len(g) for g in groups]}")
```

## Methodischer Hinweis

Die Hitzewellen-Definition (DWD: mind. 3 Tage in Folge über ortsspezifischem 98.-Perzentil-Schwellenwert **und** über 28°C) sowie Referenzperiode (1961–1990) und Analysezeitraum (1980–2025) sind unverändert aus der Deutschland-Version von RQ2 übernommen — obwohl die Definition ursprünglich vom Deutschen Wetterdienst stammt, wird sie hier bewusst **nicht** an italienische Klimaverhältnisse angepasst, damit die Ergebnisse zwischen Deutschland und Italien methodisch vergleichbar bleiben.

Die Größenklassen-Einteilung selbst weicht dagegen bewusst von der Deutschland-Version ab (siehe die "Wichtige methodische Anpassung" ganz oben): Da Italien keine dem deutschen Stadtrecht entsprechende rechtliche Unterscheidung zwischen Stadt und Gemeinde kennt, beruht die Klassifizierung hier ausschließlich auf der Einwohnerzahl, bei identischen Schwellenwerten wie in der Deutschland-Version.

Die Aggregation nach Größenklasse läuft bewusst über eindeutige Orte (`unique_places_df`/`place_level`), nicht über die volle, nicht-deduplizierte `comparison_df`-Tabelle, weil mehrere Großstädte sich denselben nächstgelegenen Vergleichsort teilen können (siehe Abschnitt 8) — ohne diese Deduplizierung würde ein einzelner, mehrfach verwendeter Ort das Ergebnis einer Größenklasse verzerren.

Vor der endgültigen Interpretation lohnt sich derselbe Blick wie in RQ2 Teil 1 empfohlen: stichprobenartig prüfen, ob die automatisch gewählten Landgemeinden tatsächlich sinnvolle ländliche Vergleichsorte sind (z. B. nicht direkt Teil eines dicht bebauten Ballungsraums).
